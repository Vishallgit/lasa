```python
import os, math, warnings
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from mlxtend.plotting import plot_decision_regions

warnings.filterwarnings('ignore')
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

from IPython.display import display, HTML
display(HTML("<style>.output_scroll {height:auto !important; max-height:none !important;}</style>"))

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)
os.environ['PYTHONHASHSEED'] = str(SEED)

print("="*60)
print("GPU CONFIGURATION")
print("="*60)
gpus = tf.config.list_physical_devices('GPU')
USE_GPU = bool(gpus)
if USE_GPU:
    print(f"GPUs available: {len(gpus)}")
    for gpu in gpus:
        print("  -", gpu)
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
        print("GPU memory growth enabled")
    except RuntimeError as e:
        print(e)
else:
    print("No GPUs available, using CPU")

policy_name = 'mixed_float16' if USE_GPU else 'float32'
tf.keras.mixed_precision.set_global_policy(policy_name)
print(f"Mixed precision policy: {tf.keras.mixed_precision.global_policy().name}")
print("="*60 + "\n")

USE_XLA = USE_GPU

def inv_softplus(y):
    y = np.maximum(float(y), 1e-6)
    return np.log(np.expm1(y))

def generate_corner_and_center_dataset(samples_per_class, noise=0.1):
    np.random.seed(SEED)
    data = []
    labels = []
    group1_corners = [(0, 0), (0, 1)]
    group2_corners = [(1, 0), (1, 1)]
    center = [(0.5, 0.5)]
    
    for corner in group1_corners:
        x1_corner, x2_corner = corner
        for _ in range(samples_per_class):
            x1 = x1_corner + np.random.normal(0, noise)
            x2 = x2_corner + np.random.normal(0, noise)
            data.append([x1, x2])
            labels.append(0)
    
    for corner in group2_corners:
        x1_corner, x2_corner = corner
        for _ in range(samples_per_class):
            x1 = x1_corner + np.random.normal(0, noise)
            x2 = x2_corner + np.random.normal(0, noise)
            data.append([x1, x2])
            labels.append(1)
    
    for center_point in center:
        x1_center, x2_center = center_point
        for _ in range(samples_per_class):
            x1 = x1_center + np.random.normal(0, noise)
            x2 = x2_center + np.random.normal(0, noise)
            data.append([x1, x2])
            labels.append(2)
    
    return np.array(data, dtype=np.float32), np.array(labels, dtype=np.int32)

class LASAActivation(tf.keras.layers.Layer):
    def __init__(self, theta_degrees=45.0, lambda_window=3e-4,
                 penalty_mode='hinge', hybrid_alpha=0.1,
                 enforce_gain_cap=False, gain_cap=1.25,
                 initial_frequency=1.6, name=None, **kwargs):
        super().__init__(name=name, **kwargs)
        self.theta_degrees     = float(theta_degrees)
        self.lambda_window     = float(lambda_window)
        self.penalty_mode      = str(penalty_mode)
        self.hybrid_alpha      = float(hybrid_alpha)
        self.enforce_gain_cap  = bool(enforce_gain_cap)
        self.gain_cap          = float(gain_cap)
        self.initial_frequency = float(initial_frequency)

    def build(self, input_shape):
        self.theta_rad = tf.constant(self.theta_degrees * math.pi/180.0, dtype=tf.float32)
        self.sin_theta = tf.sin(self.theta_rad)

        self.penalty_scale = self.add_weight(
            shape=(), name='penalty_scale', dtype=tf.float32, trainable=False,
            initializer=tf.keras.initializers.Constant(1.0)
        )

        if self.enforce_gain_cap:
            denom   = tf.maximum(self.sin_theta, tf.constant(1e-6, tf.float32))
            k_max_v = float(self.gain_cap) / float(denom.numpy())
            self.k_max = tf.constant(k_max_v, dtype=tf.float32)
            ratio = np.clip(self.initial_frequency / k_max_v, 1e-6, 1.0-1e-6)
            rho0  = np.log(ratio/(1.0 - ratio))
        else:
            self.k_max = None
            rho0 = inv_softplus(self.initial_frequency)

        self.rho = self.add_weight(shape=(), name='rho', dtype=tf.float32, trainable=True,
                                   initializer=tf.keras.initializers.Constant(float(rho0)))

    @property
    def tau_value(self):
        if self.enforce_gain_cap:
            return self.k_max * tf.nn.sigmoid(self.rho)
        return tf.nn.softplus(self.rho)

    def set_penalty_scale(self, scale):
        self.penalty_scale.assign(tf.cast(scale, tf.float32))

    def scale_down_penalty(self, factor=0.8, min_scale=0.05):
        new_scale = tf.maximum(self.penalty_scale * tf.cast(factor, tf.float32),
                               tf.constant(min_scale, tf.float32))
        self.penalty_scale.assign(new_scale)

    def _penalty(self, inputs, tau32):
        if self.lambda_window <= 0.0:
            return None
        x32     = tf.cast(inputs, tf.float32)
        cos_phi = tf.math.cos(tau32 * x32)
        abs_cos = tf.abs(cos_phi)

        if self.penalty_mode == 'hinge':
            margin = self.sin_theta
            viol   = tf.maximum(abs_cos - margin, 0.0)
            pen    = tf.reduce_mean(tf.square(viol))
        elif self.penalty_mode == 'cos2':
            pen    = tf.reduce_mean(tf.square(cos_phi))
        elif self.penalty_mode == 'hybrid':
            margin = self.sin_theta
            viol   = tf.maximum(abs_cos - margin, 0.0)
            hinge  = tf.reduce_mean(tf.square(viol))
            cos2   = tf.reduce_mean(tf.square(cos_phi))
            pen    = self.hybrid_alpha * cos2 + hinge
        else:
            raise ValueError("penalty_mode must be 'hinge', 'cos2', or 'hybrid'.")

        lw32    = tf.cast(self.lambda_window, tf.float32)
        scale32 = tf.cast(self.penalty_scale, tf.float32)
        return lw32 * scale32 * pen

    def call(self, inputs):
        tau32 = tf.cast(self.tau_value, tf.float32)
        pen = self._penalty(inputs, tau32)
        if pen is not None:
            self.add_loss(pen)
        x32 = tf.cast(inputs, tf.float32)
        y32 = tf.math.sin(tau32 * x32)
        return tf.cast(y32, inputs.dtype)

class SwishActivation(tf.keras.layers.Layer):
    def call(self, inputs):
        x = tf.cast(inputs, tf.float32)
        y = x * tf.nn.sigmoid(x)
        return tf.cast(y, inputs.dtype)

class MishActivation(tf.keras.layers.Layer):
    def call(self, inputs):
        x = tf.cast(inputs, tf.float32)
        y = x * tf.math.tanh(tf.math.softplus(x))
        return tf.cast(y, inputs.dtype)

def create_model(activation_type,
                 hidden_units=[6],
                 num_classes=3,
                 initial_frequency=1.6,
                 theta_degrees=45.0, lambda_window=3e-4,
                 penalty_mode='hinge', hybrid_alpha=0.1,
                 enforce_gain_cap=False, gain_cap=1.25):
    with tf.device('/GPU:0' if USE_GPU else '/CPU:0'):
        model = tf.keras.Sequential([tf.keras.layers.Input(shape=(2,))])
        for i, units in enumerate(hidden_units):
            model.add(tf.keras.layers.Dense(units, name=f'dense_{i}'))
            a = activation_type.lower()
            if a == 'lasa':
                model.add(LASAActivation(
                    theta_degrees=theta_degrees,
                    lambda_window=lambda_window,
                    penalty_mode=penalty_mode, hybrid_alpha=hybrid_alpha,
                    enforce_gain_cap=enforce_gain_cap, gain_cap=gain_cap,
                    initial_frequency=initial_frequency,
                    name=f'lasa_{i}'
                ))
            elif a == 'relu':
                model.add(tf.keras.layers.ReLU(name=f'relu_{i}'))
            elif a == 'leaky_relu':
                model.add(tf.keras.layers.LeakyReLU(alpha=0.01, name=f'leaky_relu_{i}'))
            elif a == 'elu':
                model.add(tf.keras.layers.ELU(alpha=1.0, name=f'elu_{i}'))
            elif a == 'softplus':
                model.add(tf.keras.layers.Activation('softplus', name=f'softplus_{i}'))
            elif a == 'swish':
                model.add(SwishActivation(name=f'swish_{i}'))
            elif a == 'mish':
                model.add(MishActivation(name=f'mish_{i}'))
            elif a == 'gelu':
                model.add(tf.keras.layers.Lambda(lambda z: tf.nn.gelu(z), name=f'gelu_{i}'))
            elif a == 'sine':
                model.add(tf.keras.layers.Lambda(lambda z: tf.math.sin(z), name=f'sine_{i}'))
            elif a == 'sigmoid':
                model.add(tf.keras.layers.Activation('sigmoid', name=f'sigmoid_{i}'))
            elif a == 'tanh':
                model.add(tf.keras.layers.Activation('tanh', name=f'tanh_{i}'))
            else:
                raise ValueError(f"Unknown activation_type: {activation_type}")

        model.add(tf.keras.layers.Dense(num_classes, activation='softmax', dtype='float32', name='output'))
    return model

class LASATrainer:
    def __init__(self, model,
                 lr_main=1e-3, lr_tau=1e-2,
                 clipnorm_main=1.0, clipnorm_tau=0.5,
                 penalty_warmup_epochs=8,
                 acc_eps=2.5e-3, loss_eps=1e-4,
                 reverse_tau_grads=True, revert_rho_to_snapshot=True,
                 shrink_tau_lr=True, tau_lr_decay=0.5, min_tau_lr=1e-5,
                 shrink_penalty=True, penalty_decay=0.8, min_penalty_scale=0.05,
                 stagnation_window=10, improve_threshold=0.05,
                 bump_delta_k=1.0, min_epochs_between_bumps=10,
                 acc_drop_threshold=0.02):
        self.model = model
        self.loss_fn = tf.keras.losses.SparseCategoricalCrossentropy()
        self.optimizer_main = tf.keras.optimizers.Adam(learning_rate=lr_main, clipnorm=clipnorm_main)
        self.optimizer_tau  = tf.keras.optimizers.Adam(learning_rate=lr_tau,  clipnorm=clipnorm_tau)
        self.penalty_warmup_epochs = int(penalty_warmup_epochs)

        self.acc_eps = float(acc_eps)
        self.loss_eps = float(loss_eps)
        self.reverse_tau_grads = bool(reverse_tau_grads)
        self.revert_rho_to_snapshot = bool(revert_rho_to_snapshot)
        self.shrink_tau_lr = bool(shrink_tau_lr)
        self.tau_lr_decay = float(tau_lr_decay)
        self.min_tau_lr = float(min_tau_lr)
        self.shrink_penalty = bool(shrink_penalty)
        self.penalty_decay = float(penalty_decay)
        self.min_penalty_scale = float(min_penalty_scale)

        self.stagnation_window = int(stagnation_window)
        self.improve_threshold = float(improve_threshold)
        self.bump_delta_k = float(bump_delta_k)
        self.min_epochs_between_bumps = int(min_epochs_between_bumps)

        self.acc_drop_threshold = float(acc_drop_threshold)

        self._partition_ready = False
        self.tau_idx, self.other_idx = [], []
        self._best_val_acc  = -np.inf
        self._best_val_loss =  np.inf
        self.val_acc_hist   = []
        self._last_bump_epoch = -10**9
        self.tau_grad_dir = tf.Variable(1.0, trainable=False, dtype=tf.float32)

        self.pending_bump = False
        self.bump_pre_val_acc = None
        self.rho_bump_snapshot = None
        self.bump_reverted_once = False
        self.final_bump_to_apply = False
        self.final_bump_applied = False
        self.disable_future_bumps = False
        self.no_revert_mode = False

    def _build_and_partition(self, sample_x):
        _ = self.model(sample_x)
        self.tau_idx, self.other_idx = [], []
        for i, v in enumerate(self.model.trainable_variables):
            if '/rho:' in v.name: self.tau_idx.append(i)
            else: self.other_idx.append(i)
        self._partition_ready = True

    def _set_penalty_scale(self, epoch_idx):
        scale = 1.0 if self.penalty_warmup_epochs <= 0 else min(1.0, (epoch_idx + 1) / self.penalty_warmup_epochs)
        for layer in self.model.layers:
            if isinstance(layer, LASAActivation):
                layer.set_penalty_scale(scale)

    def _decay_penalty_scale(self):
        if not self.shrink_penalty: return
        for layer in self.model.layers:
            if isinstance(layer, LASAActivation):
                layer.scale_down_penalty(self.penalty_decay, self.min_penalty_scale)

    def _snapshot_rho(self):
        return [self.model.trainable_variables[i].numpy().copy() for i in self.tau_idx]

    def _restore_rho(self, snap):
        for val, i in zip(snap, self.tau_idx):
            self.model.trainable_variables[i].assign(val)

    def _reset_tau_optimizer(self, new_lr):
        self.optimizer_tau = tf.keras.optimizers.Adam(learning_rate=new_lr, clipnorm=self.optimizer_tau.clipnorm)

    def _bump_all_k(self, delta):
        for layer in self.model.layers:
            if isinstance(layer, LASAActivation):
                k0 = float(layer.tau_value.numpy())
                if layer.enforce_gain_cap:
                    kmax = float(layer.k_max.numpy())
                    k1 = min(kmax * 0.999, k0 + float(delta))
                    ratio = np.clip(k1 / kmax, 1e-6, 1.0-1e-6)
                    rho_new = np.log(ratio/(1.0 - ratio))
                else:
                    k1 = max(0.0, k0 + float(delta))
                    rho_new = inv_softplus(k1)
                layer.rho.assign(np.float32(rho_new))

    def _current_k_values(self):
        ks = []
        for layer in self.model.layers:
            if isinstance(layer, LASAActivation):
                ks.append(float(layer.tau_value.numpy()))
        return ks

    @tf.function(jit_compile=USE_XLA)
    def _train_step(self, x_batch, y_batch):
        with tf.GradientTape() as tape:
            preds = self.model(x_batch, training=True)
            main_loss = self.loss_fn(y_batch, preds)
            reg_losses = tf.add_n(self.model.losses) if self.model.losses else tf.constant(0.0, dtype=main_loss.dtype)
            loss = main_loss + reg_losses
        grads = tape.gradient(loss, self.model.trainable_variables)

        tau_pairs = []
        for i in self.tau_idx:
            g = grads[i]; v = self.model.trainable_variables[i]
            if g is not None:
                g = tf.cast(g, tf.float32) * self.tau_grad_dir
                tau_pairs.append((g, v))
        other_pairs = []
        for i in self.other_idx:
            g = grads[i]; v = self.model.trainable_variables[i]
            if g is not None:
                other_pairs.append((g, v))

        if tau_pairs:   self.optimizer_tau.apply_gradients(tau_pairs)
        if other_pairs: self.optimizer_main.apply_gradients(other_pairs)

        return main_loss

    def _val_metrics(self, X_val, y_val):
        preds = self.model(X_val, training=False)
        val_loss = float(self.loss_fn(y_val, preds).numpy())
        y_pred = np.argmax(preds.numpy(), axis=1)
        val_acc = float(np.mean(y_pred == y_val.numpy()))
        return val_loss, val_acc

    def train(self, X_train, y_train, X_val, y_val, epochs=40, batch_size=32, verbose=True):
        X_train = tf.constant(X_train, dtype=tf.float32)
        y_train = tf.constant(y_train, dtype=tf.int32)
        X_val   = tf.constant(X_val,   dtype=tf.float32)
        y_val   = tf.constant(y_val,   dtype=tf.int32)

        if not self._partition_ready:
            self._build_and_partition(X_train[:1])

        ds = (tf.data.Dataset.from_tensor_slices((X_train, y_train))
              .shuffle(4096).batch(batch_size).prefetch(tf.data.AUTOTUNE))

        history = {'loss': [], 'val_accuracy': [], 'val_loss': [], 'k': []}

        for epoch in range(epochs):
            self._set_penalty_scale(epoch)

            if self.final_bump_to_apply and not self.final_bump_applied:
                self._bump_all_k(self.bump_delta_k)
                self.final_bump_applied = True
                self.final_bump_to_apply = False
                self.disable_future_bumps = True
                self.no_revert_mode = True

            rho_snap_epoch = self._snapshot_rho()

            total_loss = 0.0
            nb = 0
            for xb, yb in ds:
                main_l = self._train_step(xb, yb)
                total_loss += float(main_l.numpy())
                nb += 1

            avg_loss = total_loss / max(nb, 1)
            val_loss, val_acc = self._val_metrics(X_val, y_val)

            if self.pending_bump and not self.final_bump_applied:
                if self.bump_pre_val_acc is not None:
                    drop = self.bump_pre_val_acc - val_acc
                    if drop > self.acc_drop_threshold and not self.bump_reverted_once:
                        self._restore_rho(self.rho_bump_snapshot)
                        self.bump_reverted_once = True
                        self.pending_bump = False
                        self.final_bump_to_apply = True
                        val_loss, val_acc = self._val_metrics(X_val, y_val)
                if self.pending_bump:
                    self.pending_bump = False

            worse_acc  = (val_acc  < self._best_val_acc  - self.acc_eps)
            worse_loss = (val_loss > self._best_val_loss + self.loss_eps)
            if (worse_acc or worse_loss):
                if self.revert_rho_to_snapshot and not self.no_revert_mode:
                    self._restore_rho(rho_snap_epoch)
                if self.reverse_tau_grads:
                    self.tau_grad_dir.assign(-self.tau_grad_dir)
                if self.shrink_tau_lr:
                    new_lr = max(self.min_tau_lr, float(self.optimizer_tau.learning_rate.numpy()) * self.tau_lr_decay)
                    self._reset_tau_optimizer(new_lr)
                if self.shrink_penalty:
                    self._decay_penalty_scale()
                val_loss, val_acc = self._val_metrics(X_val, y_val)

            ep = epoch + 1
            self.val_acc_hist.append(val_acc)
            can_try_bump = (not self.disable_future_bumps) and (not self.final_bump_applied)
            if can_try_bump and ep >= self.stagnation_window and (ep - self._last_bump_epoch) >= self.min_epochs_between_bumps:
                acc_then = self.val_acc_hist[-self.stagnation_window]
                improve  = val_acc - acc_then
                if improve < self.improve_threshold:
                    self.rho_bump_snapshot = self._snapshot_rho()
                    self.bump_pre_val_acc  = val_acc
                    self._bump_all_k(self.bump_delta_k)
                    self.pending_bump = True
                    self._last_bump_epoch = ep

            self._best_val_acc  = max(self._best_val_acc,  val_acc)
            self._best_val_loss = min(self._best_val_loss, val_loss)

            history['k'].append(self._current_k_values())

            history['loss'].append(avg_loss)
            history['val_accuracy'].append(val_acc)
            history['val_loss'].append(val_loss)

            if verbose:
                print(f"Epoch {epoch+1:03d} | loss: {avg_loss:.4f} | val_loss: {val_loss:.4f} | val_acc: {val_acc:.4f}")
        return history

class RegularTrainer:
    def __init__(self, model, lr=1e-3, clipnorm=1.0):
        self.model = model
        self.loss_fn = tf.keras.losses.SparseCategoricalCrossentropy()
        self.optimizer = tf.keras.optimizers.Adam(learning_rate=lr, clipnorm=clipnorm)

    @tf.function(jit_compile=USE_XLA)
    def _train_step(self, x_batch, y_batch):
        with tf.GradientTape() as tape:
            preds = self.model(x_batch, training=True)
            loss = self.loss_fn(y_batch, preds)
        grads = tape.gradient(loss, self.model.trainable_variables)
        self.optimizer.apply_gradients(zip(grads, self.model.trainable_variables))
        return loss

    def _val_metrics(self, X_val, y_val):
        preds = self.model(X_val, training=False)
        val_loss = float(self.loss_fn(y_val, preds).numpy())
        y_pred = np.argmax(preds.numpy(), axis=1)
        val_acc = float(np.mean(y_pred == y_val.numpy()))
        return val_loss, val_acc

    def train(self, X_train, y_train, X_val, y_val, epochs=40, batch_size=32, verbose=True):
        X_train = tf.constant(X_train, dtype=tf.float32)
        y_train = tf.constant(y_train, dtype=tf.int32)
        X_val   = tf.constant(X_val,   dtype=tf.float32)
        y_val   = tf.constant(y_val,   dtype=tf.int32)

        ds = (tf.data.Dataset.from_tensor_slices((X_train, y_train))
              .shuffle(4096).batch(batch_size).prefetch(tf.data.AUTOTUNE))

        history = {'loss': [], 'val_accuracy': [], 'val_loss': []}

        for epoch in range(epochs):
            total_loss = 0.0; nb = 0
            for xb, yb in ds:
                main_l = self._train_step(xb, yb)
                total_loss += float(main_l.numpy()); nb += 1

            avg_loss = total_loss / max(nb, 1)
            val_loss, val_acc = self._val_metrics(X_val, y_val)

            history['loss'].append(avg_loss)
            history['val_accuracy'].append(val_acc)
            history['val_loss'].append(val_loss)

            if verbose:
                print(f"Epoch {epoch+1:03d} | loss: {avg_loss:.4f} | val_loss: {val_loss:.4f} | val_acc: {val_acc:.4f}")
        return history

def plot_all_decision_boundaries(models, results, X_test, y_test):
    class Wrapper:
        def __init__(self, model): self.model = model
        def predict(self, X):
            X_tensor = tf.constant(X, dtype=tf.float32)
            preds = self.model(X_tensor)
            return np.argmax(preds.numpy(), axis=1)

    names = list(models.keys())
    n = len(names); cols = 4; rows = int(np.ceil(n / cols))
    fig, axes = plt.subplots(rows, cols, figsize=(4.6*cols, 4.2*rows))
    axes = np.array(axes).reshape(rows, cols)

    for idx, name in enumerate(names):
        r, c = divmod(idx, cols); ax = axes[r, c]
        wrapper = Wrapper(models[name])
        plot_decision_regions(X=X_test, y=y_test, clf=wrapper, ax=ax, legend=0)
        acc = results[name]['accuracy']
        ax.set_title(f"{name.upper()}  Acc: {acc:.4f}", fontsize=11, fontweight='bold')
        ax.set_xlabel('x1'); ax.set_ylabel('x2'); ax.grid(True, alpha=0.3)

    for idx in range(n, rows*cols):
        r, c = divmod(idx, cols)
        fig.delaxes(axes[r, c])

    plt.suptitle('Decision Boundaries Comparison (Corner & Center Dataset)', fontsize=15, fontweight='bold')
    plt.tight_layout(); plt.show()

def plot_training_curves_comparison(results):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    for name, data in results.items():
        ax1.plot(data['history']['loss'], label=f"{name}")
        ax1.plot(data['history']['val_loss'], linestyle='--', alpha=0.7, label=f"{name} (val)")
        ax2.plot(data['history']['val_accuracy'], label=f"{name}")
    ax1.set_title('Training & Validation Loss'); ax1.set_xlabel('Epoch'); ax1.set_ylabel('Loss'); ax1.grid(True); ax1.legend(ncol=2, fontsize=8)
    ax2.set_title('Validation Accuracy'); ax2.set_xlabel('Epoch'); ax2.set_ylabel('Accuracy'); ax2.grid(True); ax2.legend(ncol=2, fontsize=8)
    plt.tight_layout(); plt.show()

def print_comparison_table(results):
    print("\n" + "="*60)
    print("FINAL RESULTS COMPARISON")
    print("="*60)
    print(f"{'Rank':<6} {'Activation':<15} {'Accuracy':<12} {'Final Train Loss':<18}")
    print("-"*60)
    sorted_items = sorted(results.items(), key=lambda kv: kv[1]['accuracy'], reverse=True)
    for rank, (name, data) in enumerate(sorted_items, 1):
        print(f"{rank:<6} {name.upper():<15} {data['accuracy']:<12.4f} {data['final_loss']:<18.4f}")
    print("="*60)

def plot_k_evolution(results, model_key='lasa', title_suffix="Corner & Center Dataset"):
    if model_key not in results:
        print(f"[plot_k_evolution] '{model_key}' not in results."); return
    hist = results[model_key]['history']
    if 'k' not in hist or len(hist['k']) == 0:
        print("[plot_k_evolution] No k history found."); return

    K = np.array([np.array(ks, dtype=float) for ks in hist['k']], dtype=object)
    try:
        K = np.stack(K).astype(float)
    except Exception:
        print("[plot_k_evolution] Ragged k history; cannot plot."); return

    plt.figure(figsize=(7, 4))
    for j in range(K.shape[1]):
        plt.plot(K[:, j], label=f'k[{j}]')
    plt.title(f'k evolution per LASA layer ({title_suffix})')
    plt.xlabel('Epoch'); plt.ylabel('k'); plt.grid(True, alpha=0.3)
    plt.legend(ncol=2)
    plt.tight_layout(); plt.show()

def run_all_activations_comparison_corner_center(hidden_units=[6],
                                                 samples_per_class=100,
                                                 noise=0.1,
                                                 test_size=0.2, random_state=42,
                                                 epochs=40, batch_size=32,
                                                 lr_main=1e-3,
                                                 lr_tau=1e-2,
                                                 initial_frequency=1.6,
                                                 theta_degrees=45.0, lambda_window=3e-4,
                                                 penalty_mode='hinge', hybrid_alpha=0.1,
                                                 enforce_gain_cap=False, gain_cap=1.25,
                                                 penalty_warmup_epochs=8,
                                                 acc_eps=2.5e-3, loss_eps=1e-4,
                                                 reverse_tau_grads=True, revert_rho_to_snapshot=True,
                                                 shrink_tau_lr=True, tau_lr_decay=0.5, min_tau_lr=1e-5,
                                                 shrink_penalty=True, penalty_decay=0.8, min_penalty_scale=0.05,
                                                 stagnation_window=10, improve_threshold=0.05,
                                                 bump_delta_k=1.0, min_epochs_between_bumps=10,
                                                 acc_drop_threshold=0.02,
                                                 verbose=True):
    activation_types = [
        'lasa', 'relu', 'leaky_relu', 'elu', 'softplus',
        'swish', 'mish', 'gelu', 'sine', 'sigmoid', 'tanh'
    ]

    print("="*60)
    print("ACTIVATION FUNCTIONS COMPARISON - Corner & Center Dataset")
    print("="*60)
    print("Config:",
          f"\n  hidden_units={hidden_units}",
          f"\n  epochs={epochs}  batch_size={batch_size}",
          f"\n  lr_main={lr_main}  (lr_tau={lr_tau} for LASA only)",
          f"\n  dataset: samples_per_class={samples_per_class}, noise={noise}",
          f"\n  LASA: theta={theta_degrees}°, lambda_window={lambda_window}, penalty_mode={penalty_mode}, "
          f"enforce_gain_cap={enforce_gain_cap}, gain_cap={gain_cap}",
          f"\n  Rule: if post-bump drop > {acc_drop_threshold:.2f}, revert-once, then one final +1 bump and disable scheduler"
          )

    X, y = generate_corner_and_center_dataset(samples_per_class, noise=noise)
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=random_state, stratify=y
    )
    
    n_samples = len(X)
    print(f"\nDataset: total={n_samples} | train={len(X_train)} | test={len(X_test)}")
    print(f"Class counts: 0->{np.sum(y==0)}  1->{np.sum(y==1)}  2->{np.sum(y==2)}")

    results = {}
    models  = {}

    for act in activation_types:
        print("\n" + "-"*60)
        print(f"Training with {act.upper()} activation...")
        tf.keras.backend.clear_session()
        np.random.seed(random_state); tf.random.set_seed(random_state)

        model = create_model(
            activation_type=act,
            hidden_units=hidden_units,
            num_classes=3,
            initial_frequency=initial_frequency,
            theta_degrees=theta_degrees, lambda_window=lambda_window,
            penalty_mode=penalty_mode, hybrid_alpha=hybrid_alpha,
            enforce_gain_cap=enforce_gain_cap, gain_cap=gain_cap
        )

        if act.lower() == 'lasa':
            trainer = LASATrainer(
                model,
                lr_main=lr_main, lr_tau=lr_tau,
                clipnorm_main=1.0, clipnorm_tau=0.5,
                penalty_warmup_epochs=penalty_warmup_epochs,
                acc_eps=acc_eps, loss_eps=loss_eps,
                reverse_tau_grads=reverse_tau_grads, revert_rho_to_snapshot=revert_rho_to_snapshot,
                shrink_tau_lr=shrink_tau_lr, tau_lr_decay=tau_lr_decay, min_tau_lr=min_tau_lr,
                shrink_penalty=shrink_penalty, penalty_decay=penalty_decay, min_penalty_scale=min_penalty_scale,
                stagnation_window=stagnation_window, improve_threshold=improve_threshold,
                bump_delta_k=bump_delta_k, min_epochs_between_bumps=min_epochs_between_bumps,
                acc_drop_threshold=acc_drop_threshold
            )
            hist = trainer.train(
                tf.constant(X_train), tf.constant(y_train, np.int32),
                tf.constant(X_test),  tf.constant(y_test,  np.int32),
                epochs=epochs, batch_size=batch_size, verbose=verbose
            )
        else:
            trainer = RegularTrainer(model, lr=lr_main, clipnorm=1.0)
            hist = trainer.train(
                tf.constant(X_train), tf.constant(y_train, np.int32),
                tf.constant(X_test),  tf.constant(y_test,  np.int32),
                epochs=epochs, batch_size=batch_size, verbose=verbose
            )

        preds = model(tf.constant(X_test, tf.float32))
        y_pred = np.argmax(preds.numpy(), axis=1)
        final_acc  = float(np.mean(y_pred == y_test))
        final_loss = float(tf.keras.losses.SparseCategoricalCrossentropy()(y_test, preds).numpy())

        results[act] = {'accuracy': final_acc, 'history': hist, 'final_loss': hist['loss'][-1]}
        models[act]  = model

        print(f"[{act.upper()}] Acc: {final_acc:.4f} | Final train loss: {hist['loss'][-1]:.4f}")

    return models, results, X, y, X_test, y_test

if __name__ == "__main__":
    models, results, X, y, X_test, y_test = run_all_activations_comparison_corner_center(
        hidden_units=[6],
        samples_per_class=100,
        noise=0.1,
        test_size=0.2, random_state=42,
        epochs=40, batch_size=32,
        lr_main=1e-3, lr_tau=1e-2,
        initial_frequency=1.6,
        theta_degrees=45.0, lambda_window=3e-4,
        penalty_mode='hinge', hybrid_alpha=0.1,
        enforce_gain_cap=False, gain_cap=1.25,
        penalty_warmup_epochs=8,
        acc_eps=2.5e-3, loss_eps=1e-4,
        reverse_tau_grads=True, revert_rho_to_snapshot=True,
        shrink_tau_lr=True, tau_lr_decay=0.5, min_tau_lr=1e-5,
        shrink_penalty=True, penalty_decay=0.8, min_penalty_scale=0.05,
        stagnation_window=10, improve_threshold=0.05,
        bump_delta_k=1.0, min_epochs_between_bumps=10,
        acc_drop_threshold=0.02,
        verbose=True
    )

    print("\n")
    print("="*60)
    print("ACTIVATIONS: lasa, relu, leaky_relu, elu, softplus, swish, mish, gelu, sine, sigmoid, tanh")
    print("="*60)
    print_comparison_table(results)
    plot_all_decision_boundaries(models, results, X_test, y_test)
    plot_training_curves_comparison(results)

    plot_k_evolution(results, model_key='lasa', title_suffix="Corner & Center Dataset")
```

```python
# ============================================================
# Environment and imports
# ============================================================
import os, math, warnings
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from mlxtend.plotting import plot_decision_regions

warnings.filterwarnings('ignore')
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

from IPython.display import display, HTML
display(HTML("<style>.output_scroll {height:auto !important; max-height:none !important;}</style>"))

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)
os.environ['PYTHONHASHSEED'] = str(SEED)

print("="*60)
print("GPU CONFIGURATION")
print("="*60)
gpus = tf.config.list_physical_devices('GPU')
USE_GPU = bool(gpus)
if USE_GPU:
    print(f"GPUs available: {len(gpus)}")
    for gpu in gpus:
        print("  -", gpu)
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
        print("GPU memory growth enabled")
    except RuntimeError as e:
        print(e)
else:
    print("No GPUs available, using CPU")

policy_name = 'mixed_float16' if USE_GPU else 'float32'
tf.keras.mixed_precision.set_global_policy(policy_name)
print(f"Mixed precision policy: {tf.keras.mixed_precision.global_policy().name}")
print("="*60 + "\n")

USE_XLA = USE_GPU  # keep XLA on only for GPU


# ============================================================
# Utilities
# ============================================================
def inv_softplus(y):
    """Inverse of softplus: given k>=0, return rho such that softplus(rho)=k."""
    y = np.maximum(float(y), 1e-6)
    return np.log(np.expm1(y))


# ============================================================
# Toy data: "corner and center" 3-class dataset
# ============================================================
def generate_corner_and_center_dataset(samples_per_class, noise=0.1):
    np.random.seed(SEED)
    data, labels = [], []
    group1_corners = [(0, 0), (0, 1)]
    group2_corners = [(1, 0), (1, 1)]
    center = [(0.5, 0.5)]

    for corner in group1_corners:
        x1_corner, x2_corner = corner
        for _ in range(samples_per_class):
            x1 = x1_corner + np.random.normal(0, noise)
            x2 = x2_corner + np.random.normal(0, noise)
            data.append([x1, x2]); labels.append(0)

    for corner in group2_corners:
        x1_corner, x2_corner = corner
        for _ in range(samples_per_class):
            x1 = x1_corner + np.random.normal(0, noise)
            x2 = x2_corner + np.random.normal(0, noise)
            data.append([x1, x2]); labels.append(1)

    for center_point in center:
        x1_center, x2_center = center_point
        for _ in range(samples_per_class):
            x1 = x1_center + np.random.normal(0, noise)
            x2 = x2_center + np.random.normal(0, noise)
            data.append([x1, x2]); labels.append(2)

    return np.array(data, dtype=np.float32), np.array(labels, dtype=np.int32)


# ============================================================
# LASA activation (with continuous dead-zone avoidance)
# ============================================================
class LASAActivation(tf.keras.layers.Layer):
    r"""
    Layer-wise Adaptive Sine Activation with phase windowing and optional dead-zone avoidance.

    Penalty modes:
      'hinge'        : penalize only when |cos(phi)| >  sin(theta)
      'cos2'         : penalize mean(cos(phi)^2)                 [attracts phases to cos=0]
      'hybrid'       : alpha*cos2 + hinge
      'band'         : two-sided hinge to keep |cos(phi)| in [cos_min, sin(theta)]
      'hybrid_band'  : alpha*cos2 + band   (use small alpha so band dominates)

    Dead-zone control (for 'band' / 'hybrid_band'):
      cos_min = deadzone_frac * sin(theta), with 0 < deadzone_frac < 1.
      This discourages |cos(phi)|=0 so slopes remain small but strictly nonzero on data.

    Frequency capping (optional):
      If enforce_gain_cap=True, tau = k_max * sigmoid(rho) with k_max = gain_cap / sin(theta).
      Otherwise tau = softplus(rho) >= 0.
    """
    def __init__(self,
                 theta_degrees=45.0,
                 lambda_window=3e-4,
                 penalty_mode='band',          # default changed to 'band'
                 hybrid_alpha=0.1,
                 enforce_gain_cap=False,
                 gain_cap=1.25,
                 initial_frequency=1.6,
                 # dead-zone band parameters
                 avoid_deadzone=True,
                 deadzone_frac=0.2,            # cos_min = deadzone_frac * sin(theta)
                 deadzone_weight=1.0,          # relative weight for lower-bound term
                 name=None, **kwargs):
        super().__init__(name=name, **kwargs)
        self.theta_degrees     = float(theta_degrees)
        self.lambda_window     = float(lambda_window)
        self.penalty_mode      = str(penalty_mode)
        self.hybrid_alpha      = float(hybrid_alpha)
        self.enforce_gain_cap  = bool(enforce_gain_cap)
        self.gain_cap          = float(gain_cap)
        self.initial_frequency = float(initial_frequency)

        self.avoid_deadzone  = bool(avoid_deadzone)
        self.deadzone_frac   = float(deadzone_frac)
        self.deadzone_weight = float(deadzone_weight)

    def build(self, input_shape):
        # constants
        self.theta_rad = tf.constant(self.theta_degrees * math.pi/180.0, dtype=tf.float32)
        self.sin_theta = tf.sin(self.theta_rad)

        # penalty scale for warm-up / decay
        self.penalty_scale = self.add_weight(
            shape=(), name='penalty_scale', dtype=tf.float32, trainable=False,
            initializer=tf.keras.initializers.Constant(1.0)
        )

        # k = tau (>=0) parameterization
        if self.enforce_gain_cap:
            denom   = tf.maximum(self.sin_theta, tf.constant(1e-6, tf.float32))
            k_max_v = float(self.gain_cap) / float(denom.numpy())
            self.k_max = tf.constant(k_max_v, dtype=tf.float32)
            ratio = np.clip(self.initial_frequency / k_max_v, 1e-6, 1.0-1e-6)
            rho0  = np.log(ratio/(1.0 - ratio))  # logit
        else:
            self.k_max = None
            rho0 = inv_softplus(self.initial_frequency)

        self.rho = self.add_weight(shape=(), name='rho', dtype=tf.float32, trainable=True,
                                   initializer=tf.keras.initializers.Constant(float(rho0)))

        # lower band for |cos| to avoid dead-zone at peaks/troughs
        if self.avoid_deadzone:
            self.cos_min = tf.cast(self.deadzone_frac, tf.float32) * self.sin_theta
        else:
            self.cos_min = tf.constant(0.0, dtype=tf.float32)

    @property
    def tau_value(self):
        if self.enforce_gain_cap:
            return self.k_max * tf.nn.sigmoid(self.rho)   # (0, k_max)
        return tf.nn.softplus(self.rho)                   # [0, ∞)

    # helpers for trainer
    def set_penalty_scale(self, scale):
        self.penalty_scale.assign(tf.cast(scale, tf.float32))

    def scale_down_penalty(self, factor=0.8, min_scale=0.05):
        new_scale = tf.maximum(self.penalty_scale * tf.cast(factor, tf.float32),
                               tf.constant(min_scale, tf.float32))
        self.penalty_scale.assign(new_scale)

    # penalty
    def _penalty(self, inputs, tau32):
        if self.lambda_window <= 0.0:
            return None
        x32     = tf.cast(inputs, tf.float32)
        cos_phi = tf.math.cos(tau32 * x32)
        abs_cos = tf.abs(cos_phi)

        mode = self.penalty_mode.lower()

        if mode == 'hinge':
            margin_hi = self.sin_theta
            viol_hi   = tf.nn.relu(abs_cos - margin_hi)
            pen       = tf.reduce_mean(tf.square(viol_hi))

        elif mode == 'cos2':
            pen = tf.reduce_mean(tf.square(cos_phi))  # attracts to cos=0

        elif mode == 'hybrid':
            margin_hi = self.sin_theta
            viol_hi   = tf.nn.relu(abs_cos - margin_hi)
            hinge     = tf.reduce_mean(tf.square(viol_hi))
            cos2      = tf.reduce_mean(tf.square(cos_phi))
            pen       = self.hybrid_alpha * cos2 + hinge

        elif mode in ('band', 'hybrid_band'):
            margin_hi = self.sin_theta
            margin_lo = self.cos_min if self.avoid_deadzone else tf.constant(0.0, tf.float32)

            # two-sided hinge: keep |cos| in [margin_lo, margin_hi]
            viol_hi = tf.nn.relu(abs_cos - margin_hi)                      # too large slope
            viol_lo = tf.nn.relu(margin_lo - abs_cos)                      # too small slope (dead-zone)
            band    = tf.reduce_mean(tf.square(viol_hi) +
                                     self.deadzone_weight * tf.square(viol_lo))

            if mode == 'hybrid_band':
                cos2 = tf.reduce_mean(tf.square(cos_phi))
                pen  = self.hybrid_alpha * cos2 + band
            else:
                pen  = band

        else:
            raise ValueError("penalty_mode must be one of "
                             "{'hinge','cos2','hybrid','band','hybrid_band'}.")

        lw32    = tf.cast(self.lambda_window, tf.float32)
        scale32 = tf.cast(self.penalty_scale, tf.float32)
        return lw32 * scale32 * pen

    def call(self, inputs):
        tau32 = tf.cast(self.tau_value, tf.float32)
        pen = self._penalty(inputs, tau32)
        if pen is not None:
            self.add_loss(pen)
        x32 = tf.cast(inputs, tf.float32)
        y32 = tf.math.sin(tau32 * x32)
        return tf.cast(y32, inputs.dtype)


# ============================================================
# Other activations for comparison
# ============================================================
class SwishActivation(tf.keras.layers.Layer):
    def call(self, inputs):
        x = tf.cast(inputs, tf.float32)
        y = x * tf.nn.sigmoid(x)
        return tf.cast(y, inputs.dtype)

class MishActivation(tf.keras.layers.Layer):
    def call(self, inputs):
        x = tf.cast(inputs, tf.float32)
        y = x * tf.math.tanh(tf.math.softplus(x))
        return tf.cast(y, inputs.dtype)


# ============================================================
# Model builder
# ============================================================
def create_model(activation_type,
                 hidden_units=[6],
                 num_classes=3,
                 initial_frequency=1.6,
                 theta_degrees=45.0, lambda_window=3e-4,
                 penalty_mode='band', hybrid_alpha=0.1,
                 enforce_gain_cap=False, gain_cap=1.25,
                 # new LASA args
                 avoid_deadzone=True, deadzone_frac=0.2, deadzone_weight=1.0):
    with tf.device('/GPU:0' if USE_GPU else '/CPU:0'):
        model = tf.keras.Sequential([tf.keras.layers.Input(shape=(2,))])
        for i, units in enumerate(hidden_units):
            model.add(tf.keras.layers.Dense(units, name=f'dense_{i}'))
            a = activation_type.lower()
            if a == 'lasa':
                model.add(LASAActivation(
                    theta_degrees=theta_degrees,
                    lambda_window=lambda_window,
                    penalty_mode=penalty_mode, hybrid_alpha=hybrid_alpha,
                    enforce_gain_cap=enforce_gain_cap, gain_cap=gain_cap,
                    initial_frequency=initial_frequency,
                    avoid_deadzone=avoid_deadzone,
                    deadzone_frac=deadzone_frac,
                    deadzone_weight=deadzone_weight,
                    name=f'lasa_{i}'
                ))
            elif a == 'relu':
                model.add(tf.keras.layers.ReLU(name=f'relu_{i}'))
            elif a == 'leaky_relu':
                model.add(tf.keras.layers.LeakyReLU(alpha=0.01, name=f'leaky_relu_{i}'))
            elif a == 'elu':
                model.add(tf.keras.layers.ELU(alpha=1.0, name=f'elu_{i}'))
            elif a == 'softplus':
                model.add(tf.keras.layers.Activation('softplus', name=f'softplus_{i}'))
            elif a == 'swish':
                model.add(SwishActivation(name=f'swish_{i}'))
            elif a == 'mish':
                model.add(MishActivation(name=f'mish_{i}'))
            elif a == 'gelu':
                model.add(tf.keras.layers.Lambda(lambda z: tf.nn.gelu(z), name=f'gelu_{i}'))
            elif a == 'sine':
                model.add(tf.keras.layers.Lambda(lambda z: tf.math.sin(z), name=f'sine_{i}'))
            elif a == 'sigmoid':
                model.add(tf.keras.layers.Activation('sigmoid', name=f'sigmoid_{i}'))
            elif a == 'tanh':
                model.add(tf.keras.layers.Activation('tanh', name=f'tanh_{i}'))
            else:
                raise ValueError(f"Unknown activation_type: {activation_type}")

        model.add(tf.keras.layers.Dense(num_classes, activation='softmax', dtype='float32', name='output'))
    return model


# ============================================================
# LASA Trainer
# ============================================================
class LASATrainer:
    """
    LASA trainer with:
      • validation guard (silent),
      • stagnation bump (+1 to k),
      • SAFETY RULE: if next-epoch val_acc drop > acc_drop_threshold vs pre-bump, revert k ONCE,
                     then one final +1 bump on the following epoch and disable future bumps/reverts.
        (k still learns by gradient; only the scheduler is disabled.)
    """
    def __init__(self, model,
                 lr_main=1e-3, lr_tau=1e-2,
                 clipnorm_main=1.0, clipnorm_tau=0.5,
                 penalty_warmup_epochs=8,
                 # validation guard
                 acc_eps=2.5e-3, loss_eps=1e-4,
                 reverse_tau_grads=True, revert_rho_to_snapshot=True,
                 shrink_tau_lr=True, tau_lr_decay=0.5, min_tau_lr=1e-5,
                 shrink_penalty=True, penalty_decay=0.8, min_penalty_scale=0.05,
                 # stagnation bump
                 stagnation_window=10, improve_threshold=0.05,
                 bump_delta_k=1.0, min_epochs_between_bumps=10,
                 # safety rule
                 acc_drop_threshold=0.02):
        self.model = model
        self.loss_fn = tf.keras.losses.SparseCategoricalCrossentropy()
        self.optimizer_main = tf.keras.optimizers.Adam(learning_rate=lr_main, clipnorm=clipnorm_main)
        self.optimizer_tau  = tf.keras.optimizers.Adam(learning_rate=lr_tau,  clipnorm=clipnorm_tau)
        self.penalty_warmup_epochs = int(penalty_warmup_epochs)

        self.acc_eps = float(acc_eps)
        self.loss_eps = float(loss_eps)
        self.reverse_tau_grads = bool(reverse_tau_grads)
        self.revert_rho_to_snapshot = bool(revert_rho_to_snapshot)
        self.shrink_tau_lr = bool(shrink_tau_lr)
        self.tau_lr_decay = float(tau_lr_decay)
        self.min_tau_lr = float(min_tau_lr)
        self.shrink_penalty = bool(shrink_penalty)
        self.penalty_decay = float(penalty_decay)
        self.min_penalty_scale = float(min_penalty_scale)

        self.stagnation_window = int(stagnation_window)
        self.improve_threshold = float(improve_threshold)
        self.bump_delta_k = float(bump_delta_k)
        self.min_epochs_between_bumps = int(min_epochs_between_bumps)

        self.acc_drop_threshold = float(acc_drop_threshold)

        # state
        self._partition_ready = False
        self.tau_idx, self.other_idx = [], []
        self._best_val_acc  = -np.inf
        self._best_val_loss =  np.inf
        self.val_acc_hist   = []
        self._last_bump_epoch = -10**9
        self.tau_grad_dir = tf.Variable(1.0, trainable=False, dtype=tf.float32)

        # bump/revert controller
        self.pending_bump = False
        self.bump_pre_val_acc = None
        self.rho_bump_snapshot = None
        self.bump_reverted_once = False
        self.final_bump_to_apply = False
        self.final_bump_applied = False
        self.disable_future_bumps = False
        self.no_revert_mode = False

    def _build_and_partition(self, sample_x):
        _ = self.model(sample_x)  # build
        self.tau_idx, self.other_idx = [], []
        for i, v in enumerate(self.model.trainable_variables):
            if '/rho:' in v.name: self.tau_idx.append(i)
            else: self.other_idx.append(i)
        self._partition_ready = True

    def _set_penalty_scale(self, epoch_idx):
        scale = 1.0 if self.penalty_warmup_epochs <= 0 else min(1.0, (epoch_idx + 1) / self.penalty_warmup_epochs)
        for layer in self.model.layers:
            if isinstance(layer, LASAActivation):
                layer.set_penalty_scale(scale)

    def _decay_penalty_scale(self):
        if not self.shrink_penalty: return
        for layer in self.model.layers:
            if isinstance(layer, LASAActivation):
                layer.scale_down_penalty(self.penalty_decay, self.min_penalty_scale)

    def _snapshot_rho(self):
        return [self.model.trainable_variables[i].numpy().copy() for i in self.tau_idx]

    def _restore_rho(self, snap):
        for val, i in zip(snap, self.tau_idx):
            self.model.trainable_variables[i].assign(val)

    def _reset_tau_optimizer(self, new_lr):
        self.optimizer_tau = tf.keras.optimizers.Adam(learning_rate=new_lr, clipnorm=self.optimizer_tau.clipnorm)

    def _bump_all_k(self, delta):
        # bump k by +delta (silent), respecting cap if any
        for layer in self.model.layers:
            if isinstance(layer, LASAActivation):
                k0 = float(layer.tau_value.numpy())
                if layer.enforce_gain_cap:
                    kmax = float(layer.k_max.numpy())
                    k1 = min(kmax * 0.999, k0 + float(delta))
                    ratio = np.clip(k1 / kmax, 1e-6, 1.0-1e-6)
                    rho_new = np.log(ratio/(1.0 - ratio))
                else:
                    k1 = max(0.0, k0 + float(delta))
                    rho_new = inv_softplus(k1)
                layer.rho.assign(np.float32(rho_new))

    def _current_k_values(self):
        ks = []
        for layer in self.model.layers:
            if isinstance(layer, LASAActivation):
                ks.append(float(layer.tau_value.numpy()))
        return ks

    @tf.function(jit_compile=USE_XLA)
    def _train_step(self, x_batch, y_batch):
        with tf.GradientTape() as tape:
            preds = self.model(x_batch, training=True)
            main_loss = self.loss_fn(y_batch, preds)
            reg_losses = tf.add_n(self.model.losses) if self.model.losses else tf.constant(0.0, dtype=main_loss.dtype)
            loss = main_loss + reg_losses
        grads = tape.gradient(loss, self.model.trainable_variables)

        # apply tau grads with direction multiplier
        tau_pairs = []
        for i in self.tau_idx:
            g = grads[i]; v = self.model.trainable_variables[i]
            if g is not None:
                g = tf.cast(g, tf.float32) * self.tau_grad_dir
                tau_pairs.append((g, v))
        other_pairs = []
        for i in self.other_idx:
            g = grads[i]; v = self.model.trainable_variables[i]
            if g is not None:
                other_pairs.append((g, v))

        if tau_pairs:   self.optimizer_tau.apply_gradients(tau_pairs)
        if other_pairs: self.optimizer_main.apply_gradients(other_pairs)

        return main_loss

    def _val_metrics(self, X_val, y_val):
        preds = self.model(X_val, training=False)
        val_loss = float(self.loss_fn(y_val, preds).numpy())
        y_pred = np.argmax(preds.numpy(), axis=1)
        val_acc = float(np.mean(y_pred == y_val.numpy()))
        return val_loss, val_acc

    def train(self, X_train, y_train, X_val, y_val, epochs=40, batch_size=32, verbose=True):
        X_train = tf.constant(X_train, dtype=tf.float32)
        y_train = tf.constant(y_train, dtype=tf.int32)
        X_val   = tf.constant(X_val,   dtype=tf.float32)
        y_val   = tf.constant(y_val,   dtype=tf.int32)

        if not self._partition_ready:
            self._build_and_partition(X_train[:1])

        ds = (tf.data.Dataset.from_tensor_slices((X_train, y_train))
              .shuffle(4096).batch(batch_size).prefetch(tf.data.AUTOTUNE))

        history = {'loss': [], 'val_accuracy': [], 'val_loss': [], 'k': []}

        for epoch in range(epochs):
            # warm-up scale for penalty
            self._set_penalty_scale(epoch)

            # scheduled final bump (after revert) exactly once, then disable scheduler
            if self.final_bump_to_apply and not self.final_bump_applied:
                self._bump_all_k(self.bump_delta_k)
                self.final_bump_applied = True
                self.final_bump_to_apply = False
                self.disable_future_bumps = True
                self.no_revert_mode = True

            rho_snap_epoch = self._snapshot_rho()  # epoch-start snapshot for guard

            total_loss = 0.0; nb = 0
            for xb, yb in ds:
                main_l = self._train_step(xb, yb)
                total_loss += float(main_l.numpy()); nb += 1

            avg_loss = total_loss / max(nb, 1)
            val_loss, val_acc = self._val_metrics(X_val, y_val)

            # Post-bump safety check
            if self.pending_bump and not self.final_bump_applied:
                if self.bump_pre_val_acc is not None:
                    drop = self.bump_pre_val_acc - val_acc
                    if drop > self.acc_drop_threshold and not self.bump_reverted_once:
                        # revert once to pre-bump k
                        self._restore_rho(self.rho_bump_snapshot)
                        self.bump_reverted_once = True
                        self.pending_bump = False
                        # schedule one final bump next epoch; disable future bumps afterward
                        self.final_bump_to_apply = True
                        # re-evaluate metrics
                        val_loss, val_acc = self._val_metrics(X_val, y_val)
                if self.pending_bump:
                    self.pending_bump = False

            # Validation guard (silent)
            worse_acc  = (val_acc  < self._best_val_acc  - self.acc_eps)
            worse_loss = (val_loss > self._best_val_loss + self.loss_eps)
            if (worse_acc or worse_loss):
                if self.revert_rho_to_snapshot and not self.no_revert_mode:
                    self._restore_rho(rho_snap_epoch)
                if self.reverse_tau_grads:
                    self.tau_grad_dir.assign(-self.tau_grad_dir)
                if self.shrink_tau_lr:
                    new_lr = max(self.min_tau_lr, float(self.optimizer_tau.learning_rate.numpy()) * self.tau_lr_decay)
                    self._reset_tau_optimizer(new_lr)
                if self.shrink_penalty:
                    self._decay_penalty_scale()
                # re-evaluate after guard actions
                val_loss, val_acc = self._val_metrics(X_val, y_val)

            # stagnation bump
            ep = epoch + 1
            self.val_acc_hist.append(val_acc)
            can_try_bump = (not self.disable_future_bumps) and (not self.final_bump_applied)
            if can_try_bump and ep >= self.stagnation_window and (ep - self._last_bump_epoch) >= self.min_epochs_between_bumps:
                acc_then = self.val_acc_hist[-self.stagnation_window]
                improve  = val_acc - acc_then
                if improve < self.improve_threshold:
                    # snapshot pre-bump k + pre-bump acc
                    self.rho_bump_snapshot = self._snapshot_rho()
                    self.bump_pre_val_acc  = val_acc
                    # trial bump
                    self._bump_all_k(self.bump_delta_k)
                    self.pending_bump = True
                    self._last_bump_epoch = ep

            # update bests
            self._best_val_acc  = max(self._best_val_acc,  val_acc)
            self._best_val_loss = min(self._best_val_loss, val_loss)

            # record k per epoch
            history['k'].append(self._current_k_values())

            history['loss'].append(avg_loss)
            history['val_accuracy'].append(val_acc)
            history['val_loss'].append(val_loss)

            if verbose:
                print(f"Epoch {epoch+1:03d} | loss: {avg_loss:.4f} | val_loss: {val_loss:.4f} | val_acc: {val_acc:.4f}")
        return history


# ============================================================
# Baseline trainer (for non-LASA activations)
# ============================================================
class RegularTrainer:
    def __init__(self, model, lr=1e-3, clipnorm=1.0):
        self.model = model
        self.loss_fn = tf.keras.losses.SparseCategoricalCrossentropy()
        self.optimizer = tf.keras.optimizers.Adam(learning_rate=lr, clipnorm=clipnorm)

    @tf.function(jit_compile=USE_XLA)
    def _train_step(self, x_batch, y_batch):
        with tf.GradientTape() as tape:
            preds = self.model(x_batch, training=True)
            loss = self.loss_fn(y_batch, preds)
        grads = tape.gradient(loss, self.model.trainable_variables)
        self.optimizer.apply_gradients(zip(grads, self.model.trainable_variables))
        return loss

    def _val_metrics(self, X_val, y_val):
        preds = self.model(X_val, training=False)
        val_loss = float(self.loss_fn(y_val, preds).numpy())
        y_pred = np.argmax(preds.numpy(), axis=1)
        val_acc = float(np.mean(y_pred == y_val.numpy()))
        return val_loss, val_acc

    def train(self, X_train, y_train, X_val, y_val, epochs=40, batch_size=32, verbose=True):
        X_train = tf.constant(X_train, dtype=tf.float32)
        y_train = tf.constant(y_train, dtype=tf.int32)
        X_val   = tf.constant(X_val,   dtype=tf.float32)
        y_val   = tf.constant(y_val,   dtype=tf.int32)

        ds = (tf.data.Dataset.from_tensor_slices((X_train, y_train))
              .shuffle(4096).batch(batch_size).prefetch(tf.data.AUTOTUNE))

        history = {'loss': [], 'val_accuracy': [], 'val_loss': []}

        for epoch in range(epochs):
            total_loss = 0.0; nb = 0
            for xb, yb in ds:
                main_l = self._train_step(xb, yb)
                total_loss += float(main_l.numpy()); nb += 1

            avg_loss = total_loss / max(nb, 1)
            val_loss, val_acc = self._val_metrics(X_val, y_val)

            history['loss'].append(avg_loss)
            history['val_accuracy'].append(val_acc)
            history['val_loss'].append(val_loss)

            if verbose:
                print(f"Epoch {epoch+1:03d} | loss: {avg_loss:.4f} | val_loss: {val_loss:.4f} | val_acc: {val_acc:.4f}")
        return history


# ============================================================
# Visualisation helpers
# ============================================================
def plot_all_decision_boundaries(models, results, X_test, y_test):
    class Wrapper:
        def __init__(self, model): self.model = model
        def predict(self, X):
            X_tensor = tf.constant(X, dtype=tf.float32)
            preds = self.model(X_tensor)
            return np.argmax(preds.numpy(), axis=1)

    names = list(models.keys())
    n = len(names); cols = 4; rows = int(np.ceil(n / cols))
    fig, axes = plt.subplots(rows, cols, figsize=(4.6*cols, 4.2*rows))
    axes = np.array(axes).reshape(rows, cols)

    for idx, name in enumerate(names):
        r, c = divmod(idx, cols); ax = axes[r, c]
        wrapper = Wrapper(models[name])
        plot_decision_regions(X=X_test, y=y_test, clf=wrapper, ax=ax, legend=0)
        acc = results[name]['accuracy']
        ax.set_title(f"{name.upper()}  Acc: {acc:.4f}", fontsize=11, fontweight='bold')
        ax.set_xlabel('x1'); ax.set_ylabel('x2'); ax.grid(True, alpha=0.3)

    for idx in range(n, rows*cols):
        r, c = divmod(idx, cols)
        fig.delaxes(axes[r, c])

    plt.suptitle('Decision Boundaries Comparison (Corner & Center Dataset)', fontsize=15, fontweight='bold')
    plt.tight_layout(); plt.show()

def plot_training_curves_comparison(results):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    for name, data in results.items():
        ax1.plot(data['history']['loss'], label=f"{name}")
        ax1.plot(data['history']['val_loss'], linestyle='--', alpha=0.7, label=f"{name} (val)")
        ax2.plot(data['history']['val_accuracy'], label=f"{name}")
    ax1.set_title('Training & Validation Loss'); ax1.set_xlabel('Epoch'); ax1.set_ylabel('Loss'); ax1.grid(True); ax1.legend(ncol=2, fontsize=8)
    ax2.set_title('Validation Accuracy'); ax2.set_xlabel('Epoch'); ax2.set_ylabel('Accuracy'); ax2.grid(True); ax2.legend(ncol=2, fontsize=8)
    plt.tight_layout(); plt.show()

def print_comparison_table(results):
    print("\n" + "="*60)
    print("FINAL RESULTS COMPARISON")
    print("="*60)
    print(f"{'Rank':<6} {'Activation':<15} {'Accuracy':<12} {'Final Train Loss':<18}")
    print("-"*60)
    sorted_items = sorted(results.items(), key=lambda kv: kv[1]['accuracy'], reverse=True)
    for rank, (name, data) in enumerate(sorted_items, 1):
        print(f"{rank:<6} {name.upper():<15} {data['accuracy']:<12.4f} {data['final_loss']:<18.4f}")
    print("="*60)

def plot_k_evolution(results, model_key='lasa', title_suffix="Corner & Center Dataset"):
    if model_key not in results:
        print(f"[plot_k_evolution] '{model_key}' not in results."); return
    hist = results[model_key]['history']
    if 'k' not in hist or len(hist['k']) == 0:
        print("[plot_k_evolution] No k history found."); return

    K = np.array([np.array(ks, dtype=float) for ks in hist['k']], dtype=object)
    try:
        K = np.stack(K).astype(float)  # shape: (E, L)
    except Exception:
        print("[plot_k_evolution] Ragged k history; cannot plot."); return

    plt.figure(figsize=(7, 4))
    for j in range(K.shape[1]):
        plt.plot(K[:, j], label=f'k[{j}]')
    plt.title(f'k evolution per LASA layer ({title_suffix})')
    plt.xlabel('Epoch'); plt.ylabel('k'); plt.grid(True, alpha=0.3)
    plt.legend(ncol=2)
    plt.tight_layout(); plt.show()


# ============================================================
# Experiment runner
# ============================================================
def run_all_activations_comparison_corner_center(hidden_units=[6],
                                                 samples_per_class=100,
                                                 noise=0.1,
                                                 test_size=0.2, random_state=42,
                                                 epochs=40, batch_size=32,
                                                 lr_main=1e-3,
                                                 lr_tau=1e-2,
                                                 initial_frequency=1.6,
                                                 theta_degrees=45.0, lambda_window=3e-4,
                                                 penalty_mode='band', hybrid_alpha=0.1,
                                                 enforce_gain_cap=False, gain_cap=1.25,
                                                 # LASA-specific
                                                 avoid_deadzone=True, deadzone_frac=0.2, deadzone_weight=1.0,
                                                 penalty_warmup_epochs=8,
                                                 # guards/controller
                                                 acc_eps=2.5e-3, loss_eps=1e-4,
                                                 reverse_tau_grads=True, revert_rho_to_snapshot=True,
                                                 shrink_tau_lr=True, tau_lr_decay=0.5, min_tau_lr=1e-5,
                                                 shrink_penalty=True, penalty_decay=0.8, min_penalty_scale=0.05,
                                                 stagnation_window=10, improve_threshold=0.05,
                                                 bump_delta_k=1.0, min_epochs_between_bumps=10,
                                                 # safety threshold
                                                 acc_drop_threshold=0.02,
                                                 verbose=True):
    activation_types = [
        'lasa', 'relu', 'leaky_relu', 'elu', 'softplus',
        'swish', 'mish', 'gelu', 'sine', 'sigmoid', 'tanh'
    ]

    print("="*60)
    print("ACTIVATION FUNCTIONS COMPARISON - Corner & Center Dataset")
    print("="*60)
    print("Config:",
          f"\n  hidden_units={hidden_units}",
          f"\n  epochs={epochs}  batch_size={batch_size}",
          f"\n  lr_main={lr_main}  (lr_tau={lr_tau} for LASA only)",
          f"\n  dataset: samples_per_class={samples_per_class}, noise={noise}",
          f"\n  LASA: theta={theta_degrees}°, lambda_window={lambda_window}, penalty_mode={penalty_mode}, "
          f"avoid_deadzone={avoid_deadzone}, deadzone_frac={deadzone_frac}, deadzone_weight={deadzone_weight}, "
          f"enforce_gain_cap={enforce_gain_cap}, gain_cap={gain_cap}",
          f"\n  Rule: if post-bump drop > {acc_drop_threshold:.2f}, revert-once, then one final +1 bump and disable scheduler"
          )

    # data
    X, y = generate_corner_and_center_dataset(samples_per_class, noise=noise)
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=random_state, stratify=y
    )

    n_samples = len(X)
    print(f"\nDataset: total={n_samples} | train={len(X_train)} | test={len(X_test)}")
    print(f"Class counts: 0->{np.sum(y==0)}  1->{np.sum(y==1)}  2->{np.sum(y==2)}")

    results = {}
    models  = {}

    for act in activation_types:
        print("\n" + "-"*60)
        print(f"Training with {act.upper()} activation...")
        tf.keras.backend.clear_session()
        np.random.seed(random_state); tf.random.set_seed(random_state)

        model = create_model(
            activation_type=act,
            hidden_units=hidden_units,
            num_classes=3,
            initial_frequency=initial_frequency,
            theta_degrees=theta_degrees, lambda_window=lambda_window,
            penalty_mode=penalty_mode, hybrid_alpha=hybrid_alpha,
            enforce_gain_cap=enforce_gain_cap, gain_cap=gain_cap,
            avoid_deadzone=avoid_deadzone, deadzone_frac=deadzone_frac, deadzone_weight=deadzone_weight
        )

        if act.lower() == 'lasa':
            trainer = LASATrainer(
                model,
                lr_main=lr_main, lr_tau=lr_tau,
                clipnorm_main=1.0, clipnorm_tau=0.5,
                penalty_warmup_epochs=penalty_warmup_epochs,
                acc_eps=acc_eps, loss_eps=loss_eps,
                reverse_tau_grads=reverse_tau_grads, revert_rho_to_snapshot=revert_rho_to_snapshot,
                shrink_tau_lr=shrink_tau_lr, tau_lr_decay=tau_lr_decay, min_tau_lr=min_tau_lr,
                shrink_penalty=shrink_penalty, penalty_decay=penalty_decay, min_penalty_scale=min_penalty_scale,
                stagnation_window=stagnation_window, improve_threshold=improve_threshold,
                bump_delta_k=bump_delta_k, min_epochs_between_bumps=min_epochs_between_bumps,
                acc_drop_threshold=acc_drop_threshold
            )
            hist = trainer.train(
                tf.constant(X_train), tf.constant(y_train, np.int32),
                tf.constant(X_test),  tf.constant(y_test,  np.int32),
                epochs=epochs, batch_size=batch_size, verbose=verbose
            )
        else:
            trainer = RegularTrainer(model, lr=lr_main, clipnorm=1.0)
            hist = trainer.train(
                tf.constant(X_train), tf.constant(y_train, np.int32),
                tf.constant(X_test),  tf.constant(y_test,  np.int32),
                epochs=epochs, batch_size=batch_size, verbose=verbose
            )

        preds = model(tf.constant(X_test, tf.float32))
        y_pred = np.argmax(preds.numpy(), axis=1)
        final_acc  = float(np.mean(y_pred == y_test))
        final_loss = float(tf.keras.losses.SparseCategoricalCrossentropy()(y_test, preds).numpy())

        results[act] = {'accuracy': final_acc, 'history': hist, 'final_loss': hist['loss'][-1] if 'loss' in hist else np.nan}
        models[act]  = model

        print(f"[{act.upper()}] Acc: {final_acc:.4f} | Final train loss: {hist['loss'][-1]:.4f}" if 'loss' in hist else
              f"[{act.upper()}] Acc: {final_acc:.4f}")

    return models, results, X, y, X_test, y_test


# ============================================================
# Main
# ============================================================
if __name__ == "__main__":
    models, results, X, y, X_test, y_test = run_all_activations_comparison_corner_center(
        hidden_units=[6],
        samples_per_class=100,
        noise=0.1,
        test_size=0.2, random_state=42,
        epochs=40, batch_size=32,
        lr_main=1e-3, lr_tau=1e-2,
        initial_frequency=1.6,
        theta_degrees=45.0, lambda_window=3e-4,
        penalty_mode='band', hybrid_alpha=0.05,   # band or hybrid_band (keep alpha small if hybrid)
        enforce_gain_cap=False, gain_cap=1.25,
        avoid_deadzone=True, deadzone_frac=0.2, deadzone_weight=1.0,
        penalty_warmup_epochs=8,
        acc_eps=2.5e-3, loss_eps=1e-4,
        reverse_tau_grads=True, revert_rho_to_snapshot=True,
        shrink_tau_lr=True, tau_lr_decay=0.5, min_tau_lr=1e-5,
        shrink_penalty=True, penalty_decay=0.8, min_penalty_scale=0.05,
        stagnation_window=10, improve_threshold=0.05,
        bump_delta_k=1.0, min_epochs_between_bumps=10,
        acc_drop_threshold=0.02,
        verbose=True
    )

    print("\n")
    print("="*60)
    print("ACTIVATIONS: lasa, relu, leaky_relu, elu, softplus, swish, mish, gelu, sine, sigmoid, tanh")
    print("="*60)
    print_comparison_table(results)
    plot_all_decision_boundaries(models, results, X_test, y_test)
    plot_training_curves_comparison(results)

    plot_k_evolution(results, model_key='lasa', title_suffix="Corner & Center Dataset")
```

```python
import numpy as np
import tensorflow as tf
import math

import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split


def inv_softplus(y):
    y = np.maximum(float(y), 1e-6)
    return np.log(np.expm1(y))


class LASAActivation(tf.keras.layers.Layer):
    r"""
    Layer-wise Adaptive Sine Activation with phase windowing and optional dead-zone avoidance.

    Penalty modes:
      'hinge'        : penalize only when |cos(phi)| >  sin(theta)
      'cos2'         : penalize mean(cos(phi)^2)                 [attracts phases to cos=0]
      'hybrid'       : alpha*cos2 + hinge
      'band'         : two-sided hinge to keep |cos(phi)| in [cos_min, sin(theta)]
      'hybrid_band'  : alpha*cos2 + band   (use small alpha so band dominates)

    Dead-zone control (for 'band' / 'hybrid_band'):
      cos_min = deadzone_frac * sin(theta), with 0 < deadzone_frac < 1.
      This discourages |cos(phi)|=0 so slopes remain small but strictly nonzero on data.

    Frequency capping (optional):
      If enforce_gain_cap=True, tau = k_max * sigmoid(rho) with k_max = gain_cap / sin(theta).
      Otherwise tau = softplus(rho) >= 0.
    """
    def __init__(self,
                 theta_degrees=45.0,
                 lambda_window=3e-4,
                 penalty_mode='band',          # default changed to 'band'
                 hybrid_alpha=0.1,
                 enforce_gain_cap=False,
                 gain_cap=1.25,
                 initial_frequency=1.6,
                 # dead-zone band parameters
                 avoid_deadzone=True,
                 deadzone_frac=0.2,            # cos_min = deadzone_frac * sin(theta)
                 deadzone_weight=1.0,          # relative weight for lower-bound term
                 name=None, **kwargs):
        super().__init__(name=name, **kwargs)
        self.theta_degrees     = float(theta_degrees)
        self.lambda_window     = float(lambda_window)
        self.penalty_mode      = str(penalty_mode)
        self.hybrid_alpha      = float(hybrid_alpha)
        self.enforce_gain_cap  = bool(enforce_gain_cap)
        self.gain_cap          = float(gain_cap)
        self.initial_frequency = float(initial_frequency)

        self.avoid_deadzone  = bool(avoid_deadzone)
        self.deadzone_frac   = float(deadzone_frac)
        self.deadzone_weight = float(deadzone_weight)

    def build(self, input_shape):
        # constants
        self.theta_rad = tf.constant(self.theta_degrees * math.pi/180.0, dtype=tf.float32)
        self.sin_theta = tf.sin(self.theta_rad)

        # penalty scale for warm-up / decay
        self.penalty_scale = self.add_weight(
            shape=(), name='penalty_scale', dtype=tf.float32, trainable=False,
            initializer=tf.keras.initializers.Constant(1.0)
        )

        # k = tau (>=0) parameterization
        if self.enforce_gain_cap:
            denom   = tf.maximum(self.sin_theta, tf.constant(1e-6, tf.float32))
            k_max_v = float(self.gain_cap) / float(denom.numpy())
            self.k_max = tf.constant(k_max_v, dtype=tf.float32)
            ratio = np.clip(self.initial_frequency / k_max_v, 1e-6, 1.0-1e-6)
            rho0  = np.log(ratio/(1.0 - ratio))  # logit
        else:
            self.k_max = None
            rho0 = inv_softplus(self.initial_frequency)

        self.rho = self.add_weight(shape=(), name='rho', dtype=tf.float32, trainable=True,
                                   initializer=tf.keras.initializers.Constant(float(rho0)))

        # lower band for |cos| to avoid dead-zone at peaks/troughs
        if self.avoid_deadzone:
            self.cos_min = tf.cast(self.deadzone_frac, tf.float32) * self.sin_theta
        else:
            self.cos_min = tf.constant(0.0, dtype=tf.float32)

    @property
    def tau_value(self):
        if self.enforce_gain_cap:
            return self.k_max * tf.nn.sigmoid(self.rho)   # (0, k_max)
        return tf.nn.softplus(self.rho)                   # [0, ∞)

    # helpers for trainer
    def set_penalty_scale(self, scale):
        self.penalty_scale.assign(tf.cast(scale, tf.float32))

    def scale_down_penalty(self, factor=0.8, min_scale=0.05):
        new_scale = tf.maximum(self.penalty_scale * tf.cast(factor, tf.float32),
                               tf.constant(min_scale, tf.float32))
        self.penalty_scale.assign(new_scale)

    # penalty
    def _penalty(self, inputs, tau32):
        if self.lambda_window <= 0.0:
            return None
        x32     = tf.cast(inputs, tf.float32)
        cos_phi = tf.math.cos(tau32 * x32)
        abs_cos = tf.abs(cos_phi)

        mode = self.penalty_mode.lower()

        if mode == 'hinge':
            margin_hi = self.sin_theta
            viol_hi   = tf.nn.relu(abs_cos - margin_hi)
            pen       = tf.reduce_mean(tf.square(viol_hi))

        elif mode == 'cos2':
            pen = tf.reduce_mean(tf.square(cos_phi))  # attracts to cos=0

        elif mode == 'hybrid':
            margin_hi = self.sin_theta
            viol_hi   = tf.nn.relu(abs_cos - margin_hi)
            hinge     = tf.reduce_mean(tf.square(viol_hi))
            cos2      = tf.reduce_mean(tf.square(cos_phi))
            pen       = self.hybrid_alpha * cos2 + hinge

        elif mode in ('band', 'hybrid_band'):
            margin_hi = self.sin_theta
            margin_lo = self.cos_min if self.avoid_deadzone else tf.constant(0.0, tf.float32)

            # two-sided hinge: keep |cos| in [margin_lo, margin_hi]
            viol_hi = tf.nn.relu(abs_cos - margin_hi)                      # too large slope
            viol_lo = tf.nn.relu(margin_lo - abs_cos)                      # too small slope (dead-zone)
            band    = tf.reduce_mean(tf.square(viol_hi) +
                                     self.deadzone_weight * tf.square(viol_lo))

            if mode == 'hybrid_band':
                cos2 = tf.reduce_mean(tf.square(cos_phi))
                pen  = self.hybrid_alpha * cos2 + band
            else:
                pen  = band

        else:
            raise ValueError("penalty_mode must be one of "
                             "{'hinge','cos2','hybrid','band','hybrid_band'}.")

        lw32    = tf.cast(self.lambda_window, tf.float32)
        scale32 = tf.cast(self.penalty_scale, tf.float32)
        return lw32 * scale32 * pen

    def call(self, inputs):
        tau32 = tf.cast(self.tau_value, tf.float32)
        pen = self._penalty(inputs, tau32)
        if pen is not None:
            self.add_loss(pen)
        x32 = tf.cast(inputs, tf.float32)
        y32 = tf.math.sin(tau32 * x32)
        return tf.cast(y32, inputs.dtype)


class LASATrainer:
    """
    LASA trainer with:
      • validation guard (silent),
      • stagnation bump (+1 to k),
      • SAFETY RULE: if next-epoch val_acc drop > acc_drop_threshold vs pre-bump, revert k ONCE,
                     then one final +1 bump on the following epoch and disable future bumps/reverts.
        (k still learns by gradient; only the scheduler is disabled.)
    """
    def __init__(self, model,
                 lr_main=1e-3, lr_tau=1e-2,
                 clipnorm_main=1.0, clipnorm_tau=0.5,
                 penalty_warmup_epochs=8,
                 # validation guard
                 acc_eps=2.5e-3, loss_eps=1e-4,
                 reverse_tau_grads=True, revert_rho_to_snapshot=True,
                 shrink_tau_lr=True, tau_lr_decay=0.5, min_tau_lr=1e-5,
                 shrink_penalty=True, penalty_decay=0.8, min_penalty_scale=0.05,
                 # stagnation bump
                 stagnation_window=10, improve_threshold=0.05,
                 bump_delta_k=1.0, min_epochs_between_bumps=10,
                 # safety rule
                 acc_drop_threshold=0.02,
                 # XLA compilation flag
                 use_xla=False):
        self.model = model
        self.loss_fn = tf.keras.losses.SparseCategoricalCrossentropy()
        self.optimizer_main = tf.keras.optimizers.Adam(learning_rate=lr_main, clipnorm=clipnorm_main)
        self.optimizer_tau  = tf.keras.optimizers.Adam(learning_rate=lr_tau,  clipnorm=clipnorm_tau)
        self.penalty_warmup_epochs = int(penalty_warmup_epochs)
        self.use_xla = use_xla

        self.acc_eps = float(acc_eps)
        self.loss_eps = float(loss_eps)
        self.reverse_tau_grads = bool(reverse_tau_grads)
        self.revert_rho_to_snapshot = bool(revert_rho_to_snapshot)
        self.shrink_tau_lr = bool(shrink_tau_lr)
        self.tau_lr_decay = float(tau_lr_decay)
        self.min_tau_lr = float(min_tau_lr)
        self.shrink_penalty = bool(shrink_penalty)
        self.penalty_decay = float(penalty_decay)
        self.min_penalty_scale = float(min_penalty_scale)

        self.stagnation_window = int(stagnation_window)
        self.improve_threshold = float(improve_threshold)
        self.bump_delta_k = float(bump_delta_k)
        self.min_epochs_between_bumps = int(min_epochs_between_bumps)

        self.acc_drop_threshold = float(acc_drop_threshold)

        # state
        self._partition_ready = False
        self.tau_idx, self.other_idx = [], []
        self._best_val_acc  = -np.inf
        self._best_val_loss =  np.inf
        self.val_acc_hist   = []
        self._last_bump_epoch = -10**9
        self.tau_grad_dir = tf.Variable(1.0, trainable=False, dtype=tf.float32)

        # bump/revert controller
        self.pending_bump = False
        self.bump_pre_val_acc = None
        self.rho_bump_snapshot = None
        self.bump_reverted_once = False
        self.final_bump_to_apply = False
        self.final_bump_applied = False
        self.disable_future_bumps = False
        self.no_revert_mode = False

    def _build_and_partition(self, sample_x):
        _ = self.model(sample_x)  # build
        self.tau_idx, self.other_idx = [], []
        for i, v in enumerate(self.model.trainable_variables):
            if '/rho:' in v.name: self.tau_idx.append(i)
            else: self.other_idx.append(i)
        self._partition_ready = True

    def _set_penalty_scale(self, epoch_idx):
        scale = 1.0 if self.penalty_warmup_epochs <= 0 else min(1.0, (epoch_idx + 1) / self.penalty_warmup_epochs)
        for layer in self.model.layers:
            if isinstance(layer, LASAActivation):
                layer.set_penalty_scale(scale)

    def _decay_penalty_scale(self):
        if not self.shrink_penalty: return
        for layer in self.model.layers:
            if isinstance(layer, LASAActivation):
                layer.scale_down_penalty(self.penalty_decay, self.min_penalty_scale)

    def _snapshot_rho(self):
        return [self.model.trainable_variables[i].numpy().copy() for i in self.tau_idx]

    def _restore_rho(self, snap):
        for val, i in zip(snap, self.tau_idx):
            self.model.trainable_variables[i].assign(val)

    def _reset_tau_optimizer(self, new_lr):
        self.optimizer_tau = tf.keras.optimizers.Adam(learning_rate=new_lr, clipnorm=self.optimizer_tau.clipnorm)

    def _bump_all_k(self, delta):
        # bump k by +delta (silent), respecting cap if any
        for layer in self.model.layers:
            if isinstance(layer, LASAActivation):
                k0 = float(layer.tau_value.numpy())
                if layer.enforce_gain_cap:
                    kmax = float(layer.k_max.numpy())
                    k1 = min(kmax * 0.999, k0 + float(delta))
                    ratio = np.clip(k1 / kmax, 1e-6, 1.0-1e-6)
                    rho_new = np.log(ratio/(1.0 - ratio))
                else:
                    k1 = max(0.0, k0 + float(delta))
                    rho_new = inv_softplus(k1)
                layer.rho.assign(np.float32(rho_new))

    def _current_k_values(self):
        ks = []
        for layer in self.model.layers:
            if isinstance(layer, LASAActivation):
                ks.append(float(layer.tau_value.numpy()))
        return ks

    def _train_step(self, x_batch, y_batch):
        # Optionally compile with XLA
        if self.use_xla:
            return self._train_step_xla(x_batch, y_batch)
        else:
            return self._train_step_regular(x_batch, y_batch)
    
    @tf.function(jit_compile=True)
    def _train_step_xla(self, x_batch, y_batch):
        return self._train_step_impl(x_batch, y_batch)
    
    @tf.function
    def _train_step_regular(self, x_batch, y_batch):
        return self._train_step_impl(x_batch, y_batch)
    
    def _train_step_impl(self, x_batch, y_batch):
        with tf.GradientTape() as tape:
            preds = self.model(x_batch, training=True)
            main_loss = self.loss_fn(y_batch, preds)
            reg_losses = tf.add_n(self.model.losses) if self.model.losses else tf.constant(0.0, dtype=main_loss.dtype)
            loss = main_loss + reg_losses
        grads = tape.gradient(loss, self.model.trainable_variables)

        # apply tau grads with direction multiplier
        tau_pairs = []
        for i in self.tau_idx:
            g = grads[i]; v = self.model.trainable_variables[i]
            if g is not None:
                g = tf.cast(g, tf.float32) * self.tau_grad_dir
                tau_pairs.append((g, v))
        other_pairs = []
        for i in self.other_idx:
            g = grads[i]; v = self.model.trainable_variables[i]
            if g is not None:
                other_pairs.append((g, v))

        if tau_pairs:   self.optimizer_tau.apply_gradients(tau_pairs)
        if other_pairs: self.optimizer_main.apply_gradients(other_pairs)

        return main_loss

    def _val_metrics(self, X_val, y_val):
        preds = self.model(X_val, training=False)
        val_loss = float(self.loss_fn(y_val, preds).numpy())
        y_pred = np.argmax(preds.numpy(), axis=1)
        val_acc = float(np.mean(y_pred == y_val.numpy()))
        return val_loss, val_acc

    def train(self, X_train, y_train, X_val, y_val, epochs=40, batch_size=32, verbose=True):
        X_train = tf.constant(X_train, dtype=tf.float32)
        y_train = tf.constant(y_train, dtype=tf.int32)
        X_val   = tf.constant(X_val,   dtype=tf.float32)
        y_val   = tf.constant(y_val,   dtype=tf.int32)

        if not self._partition_ready:
            self._build_and_partition(X_train[:1])

        ds = (tf.data.Dataset.from_tensor_slices((X_train, y_train))
              .shuffle(4096).batch(batch_size).prefetch(tf.data.AUTOTUNE))

        history = {'loss': [], 'val_accuracy': [], 'val_loss': [], 'k': []}

        for epoch in range(epochs):
            # warm-up scale for penalty
            self._set_penalty_scale(epoch)

            # scheduled final bump (after revert) exactly once, then disable scheduler
            if self.final_bump_to_apply and not self.final_bump_applied:
                self._bump_all_k(self.bump_delta_k)
                self.final_bump_applied = True
                self.final_bump_to_apply = False
                self.disable_future_bumps = True
                self.no_revert_mode = True

            rho_snap_epoch = self._snapshot_rho()  # epoch-start snapshot for guard

            total_loss = 0.0; nb = 0
            for xb, yb in ds:
                main_l = self._train_step(xb, yb)
                total_loss += float(main_l.numpy()); nb += 1

            avg_loss = total_loss / max(nb, 1)
            val_loss, val_acc = self._val_metrics(X_val, y_val)

            # Post-bump safety check
            if self.pending_bump and not self.final_bump_applied:
                if self.bump_pre_val_acc is not None:
                    drop = self.bump_pre_val_acc - val_acc
                    if drop > self.acc_drop_threshold and not self.bump_reverted_once:
                        # revert once to pre-bump k
                        self._restore_rho(self.rho_bump_snapshot)
                        self.bump_reverted_once = True
                        self.pending_bump = False
                        # schedule one final bump next epoch; disable future bumps afterward
                        self.final_bump_to_apply = True
                        # re-evaluate metrics
                        val_loss, val_acc = self._val_metrics(X_val, y_val)
                if self.pending_bump:
                    self.pending_bump = False

            # Validation guard (silent)
            worse_acc  = (val_acc  < self._best_val_acc  - self.acc_eps)
            worse_loss = (val_loss > self._best_val_loss + self.loss_eps)
            if (worse_acc or worse_loss):
                if self.revert_rho_to_snapshot and not self.no_revert_mode:
                    self._restore_rho(rho_snap_epoch)
                if self.reverse_tau_grads:
                    self.tau_grad_dir.assign(-self.tau_grad_dir)
                if self.shrink_tau_lr:
                    new_lr = max(self.min_tau_lr, float(self.optimizer_tau.learning_rate.numpy()) * self.tau_lr_decay)
                    self._reset_tau_optimizer(new_lr)
                if self.shrink_penalty:
                    self._decay_penalty_scale()
                # re-evaluate after guard actions
                val_loss, val_acc = self._val_metrics(X_val, y_val)

            # stagnation bump
            ep = epoch + 1
            self.val_acc_hist.append(val_acc)
            can_try_bump = (not self.disable_future_bumps) and (not self.final_bump_applied)
            if can_try_bump and ep >= self.stagnation_window and (ep - self._last_bump_epoch) >= self.min_epochs_between_bumps:
                acc_then = self.val_acc_hist[-self.stagnation_window]
                improve  = val_acc - acc_then
                if improve < self.improve_threshold:
                    # snapshot pre-bump k + pre-bump acc
                    self.rho_bump_snapshot = self._snapshot_rho()
                    self.bump_pre_val_acc  = val_acc
                    # trial bump
                    self._bump_all_k(self.bump_delta_k)
                    self.pending_bump = True
                    self._last_bump_epoch = ep

            # update bests
            self._best_val_acc  = max(self._best_val_acc,  val_acc)
            self._best_val_loss = min(self._best_val_loss, val_loss)

            # record k per epoch
            history['k'].append(self._current_k_values())

            history['loss'].append(avg_loss)
            history['val_accuracy'].append(val_acc)
            history['val_loss'].append(val_loss)

            if verbose:
                print(f"Epoch {epoch+1:03d} | loss: {avg_loss:.4f} | val_loss: {val_loss:.4f} | val_acc: {val_acc:.4f}")
        return history
```

```python
import numpy as np
import tensorflow as tf
import math


def inv_softplus(y):
    """Inverse of softplus: given k>=0, return rho such that softplus(rho)=k."""
    y = np.maximum(float(y), 1e-6)
    return np.log(np.expm1(y))


class LASAActivation(tf.keras.layers.Layer):
    r"""
    Layer-wise Adaptive Sine Activation with phase windowing and optional dead-zone avoidance.

    Penalty modes:
      'hinge'        : penalize only when |cos(phi)| >  sin(theta)
      'cos2'         : penalize mean(cos(phi)^2)                 [attracts phases to cos=0]
      'hybrid'       : alpha*cos2 + hinge
      'band'         : two-sided hinge to keep |cos(phi)| in [cos_min, sin(theta)]
      'hybrid_band'  : alpha*cos2 + band   (use small alpha so band dominates)

    Dead-zone control (for 'band' / 'hybrid_band'):
      cos_min = deadzone_frac * sin(theta), with 0 < deadzone_frac < 1.
      This discourages |cos(phi)|=0 so slopes remain small but strictly nonzero on data.

    Frequency capping (optional):
      If enforce_gain_cap=True, tau = k_max * sigmoid(rho) with k_max = gain_cap / sin(theta).
      Otherwise tau = softplus(rho) >= 0.
    """
    def __init__(self,
                 theta_degrees=45.0,
                 lambda_window=3e-4,
                 penalty_mode='band',          # default changed to 'band'
                 hybrid_alpha=0.1,
                 enforce_gain_cap=False,
                 gain_cap=1.25,
                 initial_frequency=1.6,
                 # dead-zone band parameters
                 avoid_deadzone=True,
                 deadzone_frac=0.2,            # cos_min = deadzone_frac * sin(theta)
                 deadzone_weight=1.0,          # relative weight for lower-bound term
                 name=None, **kwargs):
        super().__init__(name=name, **kwargs)
        self.theta_degrees     = float(theta_degrees)
        self.lambda_window     = float(lambda_window)
        self.penalty_mode      = str(penalty_mode)
        self.hybrid_alpha      = float(hybrid_alpha)
        self.enforce_gain_cap  = bool(enforce_gain_cap)
        self.gain_cap          = float(gain_cap)
        self.initial_frequency = float(initial_frequency)

        self.avoid_deadzone  = bool(avoid_deadzone)
        self.deadzone_frac   = float(deadzone_frac)
        self.deadzone_weight = float(deadzone_weight)

    def build(self, input_shape):
        # constants
        self.theta_rad = tf.constant(self.theta_degrees * math.pi/180.0, dtype=tf.float32)
        self.sin_theta = tf.sin(self.theta_rad)

        # penalty scale for warm-up / decay
        self.penalty_scale = self.add_weight(
            shape=(), name='penalty_scale', dtype=tf.float32, trainable=False,
            initializer=tf.keras.initializers.Constant(1.0)
        )

        # k = tau (>=0) parameterization
        if self.enforce_gain_cap:
            denom   = tf.maximum(self.sin_theta, tf.constant(1e-6, tf.float32))
            k_max_v = float(self.gain_cap) / float(denom.numpy())
            self.k_max = tf.constant(k_max_v, dtype=tf.float32)
            ratio = np.clip(self.initial_frequency / k_max_v, 1e-6, 1.0-1e-6)
            rho0  = np.log(ratio/(1.0 - ratio))  # logit
        else:
            self.k_max = None
            rho0 = inv_softplus(self.initial_frequency)

        self.rho = self.add_weight(shape=(), name='rho', dtype=tf.float32, trainable=True,
                                   initializer=tf.keras.initializers.Constant(float(rho0)))

        # lower band for |cos| to avoid dead-zone at peaks/troughs
        if self.avoid_deadzone:
            self.cos_min = tf.cast(self.deadzone_frac, tf.float32) * self.sin_theta
        else:
            self.cos_min = tf.constant(0.0, dtype=tf.float32)

    @property
    def tau_value(self):
        if self.enforce_gain_cap:
            return self.k_max * tf.nn.sigmoid(self.rho)   # (0, k_max)
        return tf.nn.softplus(self.rho)                   # [0, ∞)

    # helpers for trainer
    def set_penalty_scale(self, scale):
        self.penalty_scale.assign(tf.cast(scale, tf.float32))

    def scale_down_penalty(self, factor=0.8, min_scale=0.05):
        new_scale = tf.maximum(self.penalty_scale * tf.cast(factor, tf.float32),
                               tf.constant(min_scale, tf.float32))
        self.penalty_scale.assign(new_scale)

    # penalty
    def _penalty(self, inputs, tau32):
        if self.lambda_window <= 0.0:
            return None
        x32     = tf.cast(inputs, tf.float32)
        cos_phi = tf.math.cos(tau32 * x32)
        abs_cos = tf.abs(cos_phi)

        mode = self.penalty_mode.lower()

        if mode == 'hinge':
            margin_hi = self.sin_theta
            viol_hi   = tf.nn.relu(abs_cos - margin_hi)
            pen       = tf.reduce_mean(tf.square(viol_hi))

        elif mode == 'cos2':
            pen = tf.reduce_mean(tf.square(cos_phi))  # attracts to cos=0

        elif mode == 'hybrid':
            margin_hi = self.sin_theta
            viol_hi   = tf.nn.relu(abs_cos - margin_hi)
            hinge     = tf.reduce_mean(tf.square(viol_hi))
            cos2      = tf.reduce_mean(tf.square(cos_phi))
            pen       = self.hybrid_alpha * cos2 + hinge

        elif mode in ('band', 'hybrid_band'):
            margin_hi = self.sin_theta
            margin_lo = self.cos_min if self.avoid_deadzone else tf.constant(0.0, tf.float32)

            # two-sided hinge: keep |cos| in [margin_lo, margin_hi]
            viol_hi = tf.nn.relu(abs_cos - margin_hi)                      # too large slope
            viol_lo = tf.nn.relu(margin_lo - abs_cos)                      # too small slope (dead-zone)
            band    = tf.reduce_mean(tf.square(viol_hi) +
                                     self.deadzone_weight * tf.square(viol_lo))

            if mode == 'hybrid_band':
                cos2 = tf.reduce_mean(tf.square(cos_phi))
                pen  = self.hybrid_alpha * cos2 + band
            else:
                pen  = band

        else:
            raise ValueError("penalty_mode must be one of "
                             "{'hinge','cos2','hybrid','band','hybrid_band'}.")

        lw32    = tf.cast(self.lambda_window, tf.float32)
        scale32 = tf.cast(self.penalty_scale, tf.float32)
        return lw32 * scale32 * pen

    def call(self, inputs):
        tau32 = tf.cast(self.tau_value, tf.float32)
        pen = self._penalty(inputs, tau32)
        if pen is not None:
            self.add_loss(pen)
        x32 = tf.cast(inputs, tf.float32)
        y32 = tf.math.sin(tau32 * x32)
        return tf.cast(y32, inputs.dtype)


class LASATrainer:
    """
    LASA trainer with:
      • validation guard (silent),
      • stagnation bump (+1 to k),
      • SAFETY RULE: if next-epoch val_acc drop > acc_drop_threshold vs pre-bump, revert k ONCE,
                     then one final +1 bump on the following epoch and disable future bumps/reverts.
        (k still learns by gradient; only the scheduler is disabled.)
    """
    def __init__(self, model,
                 lr_main=1e-3, lr_tau=1e-2,
                 clipnorm_main=1.0, clipnorm_tau=0.5,
                 penalty_warmup_epochs=8,
                 # validation guard
                 acc_eps=2.5e-3, loss_eps=1e-4,
                 reverse_tau_grads=True, revert_rho_to_snapshot=True,
                 shrink_tau_lr=True, tau_lr_decay=0.5, min_tau_lr=1e-5,
                 shrink_penalty=True, penalty_decay=0.8, min_penalty_scale=0.05,
                 # stagnation bump
                 stagnation_window=10, improve_threshold=0.05,
                 bump_delta_k=1.0, min_epochs_between_bumps=10,
                 # safety rule
                 acc_drop_threshold=0.02,
                 # XLA compilation flag
                 use_xla=False):
        self.model = model
        self.loss_fn = tf.keras.losses.SparseCategoricalCrossentropy()
        self.optimizer_main = tf.keras.optimizers.Adam(learning_rate=lr_main, clipnorm=clipnorm_main)
        self.optimizer_tau  = tf.keras.optimizers.Adam(learning_rate=lr_tau,  clipnorm=clipnorm_tau)
        self.penalty_warmup_epochs = int(penalty_warmup_epochs)
        self.use_xla = use_xla

        self.acc_eps = float(acc_eps)
        self.loss_eps = float(loss_eps)
        self.reverse_tau_grads = bool(reverse_tau_grads)
        self.revert_rho_to_snapshot = bool(revert_rho_to_snapshot)
        self.shrink_tau_lr = bool(shrink_tau_lr)
        self.tau_lr_decay = float(tau_lr_decay)
        self.min_tau_lr = float(min_tau_lr)
        self.shrink_penalty = bool(shrink_penalty)
        self.penalty_decay = float(penalty_decay)
        self.min_penalty_scale = float(min_penalty_scale)

        self.stagnation_window = int(stagnation_window)
        self.improve_threshold = float(improve_threshold)
        self.bump_delta_k = float(bump_delta_k)
        self.min_epochs_between_bumps = int(min_epochs_between_bumps)

        self.acc_drop_threshold = float(acc_drop_threshold)

        # state
        self._partition_ready = False
        self.tau_idx, self.other_idx = [], []
        self._best_val_acc  = -np.inf
        self._best_val_loss =  np.inf
        self.val_acc_hist   = []
        self._last_bump_epoch = -10**9
        self.tau_grad_dir = tf.Variable(1.0, trainable=False, dtype=tf.float32)

        # bump/revert controller
        self.pending_bump = False
        self.bump_pre_val_acc = None
        self.rho_bump_snapshot = None
        self.bump_reverted_once = False
        self.final_bump_to_apply = False
        self.final_bump_applied = False
        self.disable_future_bumps = False
        self.no_revert_mode = False

    def _build_and_partition(self, sample_x):
        _ = self.model(sample_x)  # build
        self.tau_idx, self.other_idx = [], []
        for i, v in enumerate(self.model.trainable_variables):
            if '/rho:' in v.name: self.tau_idx.append(i)
            else: self.other_idx.append(i)
        self._partition_ready = True

    def _set_penalty_scale(self, epoch_idx):
        scale = 1.0 if self.penalty_warmup_epochs <= 0 else min(1.0, (epoch_idx + 1) / self.penalty_warmup_epochs)
        for layer in self.model.layers:
            if isinstance(layer, LASAActivation):
                layer.set_penalty_scale(scale)

    def _decay_penalty_scale(self):
        if not self.shrink_penalty: return
        for layer in self.model.layers:
            if isinstance(layer, LASAActivation):
                layer.scale_down_penalty(self.penalty_decay, self.min_penalty_scale)

    def _snapshot_rho(self):
        return [self.model.trainable_variables[i].numpy().copy() for i in self.tau_idx]

    def _restore_rho(self, snap):
        for val, i in zip(snap, self.tau_idx):
            self.model.trainable_variables[i].assign(val)

    def _reset_tau_optimizer(self, new_lr):
        self.optimizer_tau = tf.keras.optimizers.Adam(learning_rate=new_lr, clipnorm=self.optimizer_tau.clipnorm)

    def _bump_all_k(self, delta):
        # bump k by +delta (silent), respecting cap if any
        for layer in self.model.layers:
            if isinstance(layer, LASAActivation):
                k0 = float(layer.tau_value.numpy())
                if layer.enforce_gain_cap:
                    kmax = float(layer.k_max.numpy())
                    k1 = min(kmax * 0.999, k0 + float(delta))
                    ratio = np.clip(k1 / kmax, 1e-6, 1.0-1e-6)
                    rho_new = np.log(ratio/(1.0 - ratio))
                else:
                    k1 = max(0.0, k0 + float(delta))
                    rho_new = inv_softplus(k1)
                layer.rho.assign(np.float32(rho_new))

    def _current_k_values(self):
        ks = []
        for layer in self.model.layers:
            if isinstance(layer, LASAActivation):
                ks.append(float(layer.tau_value.numpy()))
        return ks

    def _train_step(self, x_batch, y_batch):
        # Optionally compile with XLA
        if self.use_xla:
            return self._train_step_xla(x_batch, y_batch)
        else:
            return self._train_step_regular(x_batch, y_batch)
    
    @tf.function(jit_compile=True)
    def _train_step_xla(self, x_batch, y_batch):
        return self._train_step_impl(x_batch, y_batch)
    
    @tf.function
    def _train_step_regular(self, x_batch, y_batch):
        return self._train_step_impl(x_batch, y_batch)
    
    def _train_step_impl(self, x_batch, y_batch):
        with tf.GradientTape() as tape:
            preds = self.model(x_batch, training=True)
            main_loss = self.loss_fn(y_batch, preds)
            reg_losses = tf.add_n(self.model.losses) if self.model.losses else tf.constant(0.0, dtype=main_loss.dtype)
            loss = main_loss + reg_losses
        grads = tape.gradient(loss, self.model.trainable_variables)

        # apply tau grads with direction multiplier
        tau_pairs = []
        for i in self.tau_idx:
            g = grads[i]; v = self.model.trainable_variables[i]
            if g is not None:
                g = tf.cast(g, tf.float32) * self.tau_grad_dir
                tau_pairs.append((g, v))
        other_pairs = []
        for i in self.other_idx:
            g = grads[i]; v = self.model.trainable_variables[i]
            if g is not None:
                other_pairs.append((g, v))

        if tau_pairs:   self.optimizer_tau.apply_gradients(tau_pairs)
        if other_pairs: self.optimizer_main.apply_gradients(other_pairs)

        return main_loss

    def _val_metrics(self, X_val, y_val):
        preds = self.model(X_val, training=False)
        val_loss = float(self.loss_fn(y_val, preds).numpy())
        y_pred = np.argmax(preds.numpy(), axis=1)
        val_acc = float(np.mean(y_pred == y_val.numpy()))
        return val_loss, val_acc

    def train(self, X_train, y_train, X_val, y_val, epochs=40, batch_size=32, verbose=True):
        X_train = tf.constant(X_train, dtype=tf.float32)
        y_train = tf.constant(y_train, dtype=tf.int32)
        X_val   = tf.constant(X_val,   dtype=tf.float32)
        y_val   = tf.constant(y_val,   dtype=tf.int32)

        if not self._partition_ready:
            self._build_and_partition(X_train[:1])

        ds = (tf.data.Dataset.from_tensor_slices((X_train, y_train))
              .shuffle(4096).batch(batch_size).prefetch(tf.data.AUTOTUNE))

        history = {'loss': [], 'val_accuracy': [], 'val_loss': [], 'k': []}

        for epoch in range(epochs):
            # warm-up scale for penalty
            self._set_penalty_scale(epoch)

            # scheduled final bump (after revert) exactly once, then disable scheduler
            if self.final_bump_to_apply and not self.final_bump_applied:
                self._bump_all_k(self.bump_delta_k)
                self.final_bump_applied = True
                self.final_bump_to_apply = False
                self.disable_future_bumps = True
                self.no_revert_mode = True

            rho_snap_epoch = self._snapshot_rho()  # epoch-start snapshot for guard

            total_loss = 0.0; nb = 0
            for xb, yb in ds:
                main_l = self._train_step(xb, yb)
                total_loss += float(main_l.numpy()); nb += 1

            avg_loss = total_loss / max(nb, 1)
            val_loss, val_acc = self._val_metrics(X_val, y_val)

            # Post-bump safety check
            if self.pending_bump and not self.final_bump_applied:
                if self.bump_pre_val_acc is not None:
                    drop = self.bump_pre_val_acc - val_acc
                    if drop > self.acc_drop_threshold and not self.bump_reverted_once:
                        # revert once to pre-bump k
                        self._restore_rho(self.rho_bump_snapshot)
                        self.bump_reverted_once = True
                        self.pending_bump = False
                        # schedule one final bump next epoch; disable future bumps afterward
                        self.final_bump_to_apply = True
                        # re-evaluate metrics
                        val_loss, val_acc = self._val_metrics(X_val, y_val)
                if self.pending_bump:
                    self.pending_bump = False

            # Validation guard (silent)
            worse_acc  = (val_acc  < self._best_val_acc  - self.acc_eps)
            worse_loss = (val_loss > self._best_val_loss + self.loss_eps)
            if (worse_acc or worse_loss):
                if self.revert_rho_to_snapshot and not self.no_revert_mode:
                    self._restore_rho(rho_snap_epoch)
                if self.reverse_tau_grads:
                    self.tau_grad_dir.assign(-self.tau_grad_dir)
                if self.shrink_tau_lr:
                    new_lr = max(self.min_tau_lr, float(self.optimizer_tau.learning_rate.numpy()) * self.tau_lr_decay)
                    self._reset_tau_optimizer(new_lr)
                if self.shrink_penalty:
                    self._decay_penalty_scale()
                # re-evaluate after guard actions
                val_loss, val_acc = self._val_metrics(X_val, y_val)

            # stagnation bump
            ep = epoch + 1
            self.val_acc_hist.append(val_acc)
            can_try_bump = (not self.disable_future_bumps) and (not self.final_bump_applied)
            if can_try_bump and ep >= self.stagnation_window and (ep - self._last_bump_epoch) >= self.min_epochs_between_bumps:
                acc_then = self.val_acc_hist[-self.stagnation_window]
                improve  = val_acc - acc_then
                if improve < self.improve_threshold:
                    # snapshot pre-bump k + pre-bump acc
                    self.rho_bump_snapshot = self._snapshot_rho()
                    self.bump_pre_val_acc  = val_acc
                    # trial bump
                    self._bump_all_k(self.bump_delta_k)
                    self.pending_bump = True
                    self._last_bump_epoch = ep

            # update bests
            self._best_val_acc  = max(self._best_val_acc,  val_acc)
            self._best_val_loss = min(self._best_val_loss, val_loss)

            # record k per epoch
            history['k'].append(self._current_k_values())

            history['loss'].append(avg_loss)
            history['val_accuracy'].append(val_acc)
            history['val_loss'].append(val_loss)

            if verbose:
                print(f"Epoch {epoch+1:03d} | loss: {avg_loss:.4f} | val_loss: {val_loss:.4f} | val_acc: {val_acc:.4f}")
        return history
    
def generate_corner_and_center_dataset(samples_per_class, noise=0.1, seed=42):
    """
    Generate a 3-class dataset:
    - Class 0: Top-left and bottom-left corners
    - Class 1: Top-right and bottom-right corners  
    - Class 2: Center
    """
    np.random.seed(seed)
    data, labels = [], []
    
    # Class 0: left corners
    group1_corners = [(0, 0), (0, 1)]
    # Class 1: right corners
    group2_corners = [(1, 0), (1, 1)]
    # Class 2: center
    center = [(0.5, 0.5)]

    for corner in group1_corners:
        x1_corner, x2_corner = corner
        for _ in range(samples_per_class):
            x1 = x1_corner + np.random.normal(0, noise)
            x2 = x2_corner + np.random.normal(0, noise)
            data.append([x1, x2])
            labels.append(0)

    for corner in group2_corners:
        x1_corner, x2_corner = corner
        for _ in range(samples_per_class):
            x1 = x1_corner + np.random.normal(0, noise)
            x2 = x2_corner + np.random.normal(0, noise)
            data.append([x1, x2])
            labels.append(1)

    for center_point in center:
        x1_center, x2_center = center_point
        for _ in range(samples_per_class):
            x1 = x1_center + np.random.normal(0, noise)
            x2 = x2_center + np.random.normal(0, noise)
            data.append([x1, x2])
            labels.append(2)

    return np.array(data, dtype=np.float32), np.array(labels, dtype=np.int32)


# ============================================================
# Create Model with LASA
# ============================================================
def create_lasa_model(hidden_units=[6], 
                      num_classes=3,
                      initial_frequency=1.6,
                      theta_degrees=45.0,
                      lambda_window=3e-4,
                      penalty_mode='band',
                      hybrid_alpha=0.1,
                      enforce_gain_cap=False,
                      gain_cap=1.25,
                      avoid_deadzone=True,
                      deadzone_frac=0.2,
                      deadzone_weight=1.0):
    """
    Create a neural network with LASA activations.
    """
    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(2,))
    ])
    
    for i, units in enumerate(hidden_units):
        # Add dense layer
        model.add(tf.keras.layers.Dense(units, name=f'dense_{i}'))
        
        # Add LASA activation
        model.add(LASAActivation(
            theta_degrees=theta_degrees,
            lambda_window=lambda_window,
            penalty_mode=penalty_mode,
            hybrid_alpha=hybrid_alpha,
            enforce_gain_cap=enforce_gain_cap,
            gain_cap=gain_cap,
            initial_frequency=initial_frequency,
            avoid_deadzone=avoid_deadzone,
            deadzone_frac=deadzone_frac,
            deadzone_weight=deadzone_weight,
            name=f'lasa_{i}'
        ))
    
    # Output layer with softmax for classification
    model.add(tf.keras.layers.Dense(num_classes, activation='softmax', 
                                    dtype='float32', name='output'))
    
    return model


# ============================================================
# Visualization Functions
# ============================================================
def plot_dataset(X, y, title="Corner & Center Dataset"):
    """Plot the dataset."""
    plt.figure(figsize=(8, 6))
    colors = ['red', 'blue', 'green']
    labels = ['Class 0 (Left)', 'Class 1 (Right)', 'Class 2 (Center)']
    
    for i in range(3):
        mask = y == i
        plt.scatter(X[mask, 0], X[mask, 1], c=colors[i], label=labels[i], alpha=0.6)
    
    plt.xlabel('x1')
    plt.ylabel('x2')
    plt.title(title)
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()


def plot_decision_boundary(model, X, y, title="LASA Decision Boundary"):
    """Plot the decision boundary of the trained model."""
    h = 0.02  # step size in the mesh
    x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
    y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5
    xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                         np.arange(y_min, y_max, h))
    
    # Predict on mesh
    mesh_input = np.c_[xx.ravel(), yy.ravel()].astype(np.float32)
    Z = model.predict(mesh_input, verbose=0)
    Z = np.argmax(Z, axis=1)
    Z = Z.reshape(xx.shape)
    
    # Plot
    plt.figure(figsize=(10, 8))
    plt.contourf(xx, yy, Z, alpha=0.4, cmap='viridis')
    
    # Plot training points
    colors = ['red', 'blue', 'green']
    labels = ['Class 0 (Left)', 'Class 1 (Right)', 'Class 2 (Center)']
    for i in range(3):
        mask = y == i
        plt.scatter(X[mask, 0], X[mask, 1], c=colors[i], label=labels[i], 
                   edgecolors='black', linewidth=0.5)
    
    plt.xlabel('x1')
    plt.ylabel('x2')
    plt.title(title)
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()


def plot_training_history(history):
    """Plot training curves."""
    fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    
    # Loss
    axes[0].plot(history['loss'], label='Train Loss')
    axes[0].plot(history['val_loss'], label='Val Loss', linestyle='--')
    axes[0].set_xlabel('Epoch')
    axes[0].set_ylabel('Loss')
    axes[0].set_title('Training & Validation Loss')
    axes[0].legend()
    axes[0].grid(True, alpha=0.3)
    
    # Accuracy
    axes[1].plot(history['val_accuracy'], label='Val Accuracy', color='green')
    axes[1].set_xlabel('Epoch')
    axes[1].set_ylabel('Accuracy')
    axes[1].set_title('Validation Accuracy')
    axes[1].legend()
    axes[1].grid(True, alpha=0.3)
    
    # k evolution (if available)
    if 'k' in history and len(history['k']) > 0:
        k_values = np.array(history['k'])
        if k_values.ndim == 2:
            for layer_idx in range(k_values.shape[1]):
                axes[2].plot(k_values[:, layer_idx], label=f'Layer {layer_idx}')
        axes[2].set_xlabel('Epoch')
        axes[2].set_ylabel('k (frequency)')
        axes[2].set_title('LASA Frequency Evolution')
        axes[2].legend()
        axes[2].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()


# ============================================================
# Main Training Script
# ============================================================
def main():
    # Set random seeds for reproducibility
    SEED = 42
    np.random.seed(SEED)
    tf.random.set_seed(SEED)
    
    print("="*60)
    print("LASA ACTIVATION TRAINING - Corner & Center Dataset")
    print("="*60)
    
    # ----------------
    # 1. Generate Data
    # ----------------
    X, y = generate_corner_and_center_dataset(
        samples_per_class=100,
        noise=0.1,
        seed=SEED
    )
    
    # Split into train/test
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, random_state=SEED, stratify=y
    )
    
    print(f"\nDataset Statistics:")
    print(f"  Total samples: {len(X)}")
    print(f"  Training samples: {len(X_train)}")
    print(f"  Test samples: {len(X_test)}")
    print(f"  Class distribution: 0={np.sum(y==0)}, 1={np.sum(y==1)}, 2={np.sum(y==2)}")
    
    # Visualize dataset
    plot_dataset(X, y)
    
    # ----------------
    # 2. Create Model
    # ----------------
    model = create_lasa_model(
        hidden_units=[6],           # Single hidden layer with 6 neurons
        num_classes=3,
        initial_frequency=1.6,
        theta_degrees=45.0,
        lambda_window=3e-4,
        penalty_mode='band',         # Use 'band' for dead-zone avoidance
        hybrid_alpha=0.05,
        enforce_gain_cap=False,
        gain_cap=1.25,
        avoid_deadzone=True,
        deadzone_frac=0.2,
        deadzone_weight=1.0
    )
    
    print(f"\nModel Architecture:")
    model.summary()
    
    # ----------------
    # 3. Create Trainer
    # ----------------
    trainer = LASATrainer(
        model,
        # Learning rates
        lr_main=1e-3,               # Learning rate for weights/biases
        lr_tau=1e-2,                # Learning rate for LASA frequencies
        clipnorm_main=1.0,
        clipnorm_tau=0.5,
        
        # Penalty warm-up
        penalty_warmup_epochs=8,
        
        # Validation guard parameters
        acc_eps=2.5e-3,
        loss_eps=1e-4,
        reverse_tau_grads=True,
        revert_rho_to_snapshot=True,
        shrink_tau_lr=True,
        tau_lr_decay=0.5,
        min_tau_lr=1e-5,
        shrink_penalty=True,
        penalty_decay=0.8,
        min_penalty_scale=0.05,
        
        # Stagnation detection
        stagnation_window=10,
        improve_threshold=0.05,
        bump_delta_k=1.0,
        min_epochs_between_bumps=10,
        
        # Safety rule
        acc_drop_threshold=0.02,
        
        # XLA compilation (set True if GPU available)
        use_xla=False
    )
    
    # ----------------
    # 4. Train Model
    # ----------------
    print(f"\nStarting Training...")
    print("-"*60)
    
    history = trainer.train(
        X_train, y_train,
        X_test, y_test,
        epochs=40,
        batch_size=32,
        verbose=True
    )
    
    # ----------------
    # 5. Evaluate Results
    # ----------------
    print("\n" + "="*60)
    print("TRAINING COMPLETE")
    print("="*60)
    
    # Final evaluation
    test_preds = model(tf.constant(X_test, dtype=tf.float32))
    test_pred_classes = np.argmax(test_preds.numpy(), axis=1)
    test_accuracy = np.mean(test_pred_classes == y_test)
    
    print(f"\nFinal Test Accuracy: {test_accuracy:.4f}")
    print(f"Final Training Loss: {history['loss'][-1]:.4f}")
    print(f"Final Validation Loss: {history['val_loss'][-1]:.4f}")
    
    # Final k values
    if 'k' in history and len(history['k']) > 0:
        final_k = history['k'][-1]
        print(f"\nFinal LASA Frequencies (k):")
        for i, k_val in enumerate(final_k):
            print(f"  Layer {i}: {k_val:.4f}")
    
    # ----------------
    # 6. Visualizations
    # ----------------
    print("\nGenerating visualizations...")
    
    # Plot training history
    plot_training_history(history)
    
    # Plot decision boundary
    plot_decision_boundary(model, X_test, y_test)
    
    return model, history


# ============================================================
# Run the example
# ============================================================
if __name__ == "__main__":
    # Make sure to include the LASA code from the previous artifact
    # either by importing it or pasting it above this section
    
    model, history = main()
```
